```python
import cartopy.crs as ccrs
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(15,10))
ax = fig.add_subplot(1,1,1, projection=ccrs.PlateCarree())
ax.stock_img()

# 父系迁徙路径（青铜时代）
paternal_path = [
    [111.0, 35.0], # 山西运城
    [109.5, 34.5], # 陕西渭南
    [112.4, 34.7], # 河南洛阳
    [112.1, 32.0] # 湖北襄阳
]
ax.plot(*zip(*paternal_path), 'r-', linewidth=2, label='O-Z26030 (4300-3500 BP)')

# 新石器核心区标注
ax.scatter(112.6, 29.3, c='blue', s=80, label='玉蟾岩遗址(9000 BP)')
ax.annotate('水稻起源', xy=(112.6,29.3), xytext=(115,30), 
            arrowprops=dict(arrowstyle='->'))

ax.legend()
plt.savefig('O-Z26030_migration.png', dpi=300)
```
