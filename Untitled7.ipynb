```python
import numpy as np
from scipy.optimize import linear_sum_assignment
%matplotlib inline
import matplotlib.pyplot as plt
import networkx as nx
import math
from itertools import combinations
import matplotlib.colors as mcolors
import time

class BellFriezeATSP:
    def __init__(self, cost_matrix):
        self.C = cost_matrix
        self.n = len(cost_matrix)
        self.ap_edges = None
        self.T_star = None
        self.non_tree_edges = None
        self.valid_cycles = None
        self.candidates = {}
        self.figures = []  # 存储图形对象
        
        # 算法参数
        self.gamma = 1.0
        self.zeta = 1.0
        self.k_max = max(1, math.ceil((math.log(self.n) ** 2.5)))  # 最大非树边数 log^{2+o(1)}n
        
    def solve_ap(self):
        """求解指派问题 (AP)"""
        row_ind, col_ind = linear_sum_assignment(self.C)
        self.ap_edges = set(zip(row_ind, col_ind))
        self.ap_cost = self.C[row_ind, col_ind].sum()
        print("ap solutions:\n")
        print(self.ap_edges)
        
        # 可视化AP解
        self.visualize_matching(self.ap_edges, "AP optimal solutions", f"cost: {self.ap_cost:.4f}")
        return self.ap_edges, self.ap_cost
    
    def build_spanning_tree(self):
        """构建生成树 T*"""
        # 创建二分图
        G = nx.Graph()
        G.add_nodes_from(range(self.n), bipartite=0, subset='A')  # A 集合
        G.add_nodes_from(range(self.n, 2*self.n), bipartite=1, subset='B')  # B 集合
        
        # 添加所有边
        for i in range(self.n):
            for j in range(self.n):
                G.add_edge(i, j + self.n, weight=self.C[i][j])
        
        # 添加AP解中的边
        for i, j in self.ap_edges:
            G.add_edge(i, j + self.n, in_ap=True, color='red', width=2.0)
        
        # 构建包含AP解的生成树
        T = nx.Graph()
        T.add_edges_from([(i, j + self.n) for i, j in self.ap_edges])
        
        # 添加边直到形成生成树
        while len(T.nodes) < 2 * self.n:
            min_edge = None
            min_cost = float('inf')
            
            for i in range(self.n):
                for j in range(self.n):
                    u, v = i, j + self.n
                    if not T.has_edge(u, v) and not T.has_node(v):
                        if self.C[i][j] < min_cost:
                            min_cost = self.C[i][j]
                            min_edge = (u, v)
            
            if min_edge:
                T.add_edge(*min_edge, weight=min_cost, color='blue', width=1.5)
            else:
                break
        
        self.T_star = T
        A = nx.to_numpy_array(T)
        print("YYC spaning tree\n")
        print(A)
        
        # 可视化生成树
        self.visualize_spanning_tree(T)
        return T
    
    def get_non_tree_edges(self):
        """识别所有非树边"""
        all_edges = set()
        for i in range(self.n):
            for j in range(self.n):
                all_edges.add((i, j))
        
        tree_edges = set()
        for u, v in self.T_star.edges():
            if u < self.n and v >= self.n:
                tree_edges.add((u, v - self.n))
        
        self.non_tree_edges = all_edges - tree_edges
        
        # 可视化非树边
        self.visualize_non_tree_edges()
        return self.non_tree_edges
    
    def find_valid_structures(self):
        """枚举所有有效环"""
        self.valid_cycles = []
        
        # 枚举所有可能的非树边组合
        for k in range(1, min(3, self.k_max) + 1):  # 限制k≤3以提高性能
            for non_tree_set in combinations(self.non_tree_edges, k):
                # 创建子图
                subG = nx.Graph()
                
                # 添加树边
                for u, v in self.T_star.edges():
                    subG.add_edge(u, v, color='blue', width=1.0)
                
                # 添加非树边
                for (i, j) in non_tree_set:
                    subG.add_edge(i, j + self.n, color='green', width=2.0)
                
                # 枚举所有简单环
                try:
                    for cycle in nx.simple_cycles(subG.to_directed()):
                        if len(cycle) % 2 == 0:  # 交替环必须偶数长度
                            edges = []
                            valid = True
                            
                            # 将节点序列转换为边序列
                            for idx in range(len(cycle)):
                                u = cycle[idx]
                                v = cycle[(idx + 1) % len(cycle)]
                                
                                if u < self.n and v >= self.n:  # A->B
                                    edge = (u, v - self.n)
                                elif u >= self.n and v < self.n:  # B->A
                                    edge = (v, u - self.n)
                                else:
                                    valid = False
                                    break
                                
                                edges.append(edge)
                            
                            if valid:
                                self.valid_cycles.append(edges)
                                
                                # 可视化有效环
                                if len(self.valid_cycles) <= 5:  # 限制可视化数量
                                    self.visualize_cycle(edges, non_tree_set)
                except nx.NetworkXNoCycle:
                    pass
        
        print(f"找到有效环: {len(self.valid_cycles)} 个")
        return self.valid_cycles
    
    def generate_candidates(self):
        """生成候选解"""
        # 原始AP解
        self.candidates[tuple(sorted(self.ap_edges))] = self.ap_cost
        
        # 添加有效环生成的候选解
        for i, cycle_edges in enumerate(self.valid_cycles):
            new_edges = set(self.ap_edges)
            
            # 应用对称差
            for edge in cycle_edges:
                if edge in new_edges:
                    new_edges.remove(edge)
                else:
                    new_edges.add(edge)
            
            # 计算成本
            cost = sum(self.C[i][j] for (i, j) in new_edges)
            self.candidates[tuple(sorted(new_edges))] = cost
            
            # 可视化候选解
            if i < 3:  # 只可视化前3个候选解
                self.visualize_matching(new_edges, f" candidates {i+1}", f" costs: {cost:.4f}")
        
        return self.candidates
    
    def find_optimal_tour(self):
        """找到最优ATSP解"""
        optimal_cost = float('inf')
        optimal_tour = None
        
        for edges, cost in self.candidates.items():
            if self.is_hamiltonian(edges) and cost < optimal_cost:
                optimal_cost = cost
                optimal_tour = edges
        
        return optimal_tour, optimal_cost
    
    def is_hamiltonian(self, edges):
        """检查边集是否形成Hamilton环"""
        graph = {i: [] for i in range(self.n)}
        for i, j in edges:
            graph[i].append(j)
        
        # 检查是否每个节点出度为1
        if any(len(graph[i]) != 1 for i in range(self.n)):
            return False
        
        # 检查连通性
        start = 0
        visited = set()
        while start not in visited:
            visited.add(start)
            start = graph[start][0]
        
        return len(visited) == self.n
    
    def visualize_matching(self, edges, title, subtitle=""):
        """可视化匹配"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 创建二分图
        G = nx.Graph()
        G.add_nodes_from(range(self.n), bipartite=0, subset='A')  # A 集合
        G.add_nodes_from(range(self.n, 2*self.n), bipartite=1, subset='B')  # B 集合
        
        # 添加所有边（灰色）
        for i in range(self.n):
            for j in range(self.n):
                G.add_edge(i, j + self.n, weight=self.C[i][j], color='lightgray', alpha=0.3)
        
        # 添加匹配边（红色）
        for i, j in edges:
            G.add_edge(i, j + self.n, color='red', width=3.0)
        
        # 布局
        pos = {}
        for i in range(self.n):
            pos[i] = (0, i)  # A侧节点在左边
        for j in range(self.n):
            pos[j + self.n] = (1, j)  # B侧节点在右边

        # 创建自定义标签：右侧节点显示原始列索引 (0 到 n-1)
        labels = {}
        for node in G.nodes():
            if node < self.n:  # A 集合节点
                labels[node] = f"{node}"  # 添加前缀 'a' 表示 A 集合
            else:  # B 集合节点
                labels[node] = f"{node - self.n}"  # 减去 n 得到原始列索引，添加前缀 'b'
        
        # 绘制
        edge_colors = [G[u][v]['color'] for u, v in G.edges()]
        edge_widths = [G[u][v].get('width', 1.0) for u, v in G.edges()]
        
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
        nx.draw_networkx_edges(G, pos, edge_color=edge_colors, width=edge_widths, alpha=0.7)
        
        # 添加标题
        plt.title(f"{title}\n{subtitle}", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_spanning_tree(self, T):
        """可视化生成树"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 布局
        pos = {}
        for i in range(self.n):
            pos[i] = (0, i)  # A侧节点在左边
        for j in range(self.n):
            pos[j + self.n] = (1, j)  # B侧节点在右边
        
        # 绘制节点
        nx.draw_networkx_nodes(T, pos, node_size=500, node_color='skyblue')
        nx.draw_networkx_labels(T, pos, font_size=10, font_weight='bold')
        
        # 绘制边（AP边红色，其他蓝色）
        ap_edges = [(i, j + self.n) for i, j in self.ap_edges]
        other_edges = [e for e in T.edges() if e not in ap_edges]
        
        nx.draw_networkx_edges(T, pos, edgelist=ap_edges, edge_color='red', width=2.5)
        nx.draw_networkx_edges(T, pos, edgelist=other_edges, edge_color='blue', width=1.5)
        
        # 添加标题
        plt.title(f"Spaning tree T* (contans AP solutions)", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_non_tree_edges(self):
        """可视化非树边"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 创建图
        G = nx.Graph()
        G.add_nodes_from(range(self.n), bipartite=0, subset='A')
        G.add_nodes_from(range(self.n, 2*self.n), bipartite=1, subset='B')
        
        # 添加树边（蓝色）
        for u, v in self.T_star.edges():
            G.add_edge(u, v, color='blue', width=1.5)
        
        # 添加非树边（绿色）
        for (i, j) in self.non_tree_edges:
            G.add_edge(i, j + self.n, color='green', width=2.0, style='dashed')
        
        # 布局
        pos = {}
        for i in range(self.n):
            pos[i] = (0, i)
        for j in range(self.n):
            pos[j + self.n] = (1, j)
        
        # 绘制
        edge_colors = [G[u][v]['color'] for u, v in G.edges()]
        edge_widths = [G[u][v].get('width', 1.0) for u, v in G.edges()]
        edge_styles = [G[u][v].get('style', 'solid') for u, v in G.edges()]
        
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
        nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
        
        for i, (u, v) in enumerate(G.edges()):
            nx.draw_networkx_edges(G, pos, edgelist=[(u, v)], 
                                  edge_color=edge_colors[i], 
                                  width=edge_widths[i],
                                  style=edge_styles[i])
        
        # 添加标题和图例
        plt.title("Spaning tree T* and non-tree edges", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_cycle(self, edges, non_tree_set):
        """可视化有效环"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 创建图
        G = nx.Graph()
        G.add_nodes_from(range(self.n), bipartite=0, subset='A')
        G.add_nodes_from(range(self.n, 2*self.n), bipartite=1, subset='B')
        
        # 添加边
        for i, j in edges:
            if (i, j) in non_tree_set:
                # 非树边（绿色虚线）
                G.add_edge(i, j + self.n, color='green', width=2.5, style='dashed')
            else:
                # 树边（蓝色实线）
                G.add_edge(i, j + self.n, color='blue', width=2.0)
        
        # 布局
        pos = {}
        for i in range(self.n):
            pos[i] = (0, i)
        for j in range(self.n):
            pos[j + self.n] = (1, j)
        
        # 绘制
        edge_colors = [G[u][v]['color'] for u, v in G.edges()]
        edge_widths = [G[u][v].get('width', 1.0) for u, v in G.edges()]
        edge_styles = [G[u][v].get('style', 'solid') for u, v in G.edges()]
        
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
        nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
        
        for i, (u, v) in enumerate(G.edges()):
            nx.draw_networkx_edges(G, pos, edgelist=[(u, v)], 
                                  edge_color=edge_colors[i], 
                                  width=edge_widths[i],
                                  style=edge_styles[i])
        
        # 添加标题
        non_tree_count = len([e for e in edges if e in self.non_tree_edges])
        plt.title(f"valid cyles (non-tree edges: {non_tree_count})", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_tour(self, tour_edges, cost):
        """可视化ATSP环"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 添加边
        for i, j in tour_edges:
            G.add_edge(i, j, weight=self.C[i][j])
        
        # 创建环形布局
        pos = nx.circular_layout(G)
        
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=700, node_color='lightgreen')
        nx.draw_networkx_labels(G, pos, font_size=12, font_weight='bold')
        
        # 绘制边
        edge_labels = {(i, j): f"{self.C[i][j]:.2f}" for i, j in G.edges()}
        nx.draw_networkx_edges(G, pos, edge_color='blue', width=2.0, 
                              arrows=True, arrowsize=20, 
                              connectionstyle='arc3,rad=0.1')
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10)
        
        # 添加标题
        plt.title(f"ATSP optimal solutions\ntotal costs: {cost:.4f}", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def show_visualizations(self):
        """显示所有可视化结果"""
        print(f"显示 {len(self.figures)} 个可视化图表...")
        plt.show()
    
    def solve(self):
        """执行完整算法"""
        print(f"解决 {self.n} 城市 ATSP 问题 (k_max = {self.k_max})")
        
        # 步骤1: 求解AP
        self.solve_ap()
        print(f"AP最优解成本: {self.ap_cost:.4f}")
        
        # 步骤2: 构建生成树T*
        self.build_spanning_tree()
        print(f"生成树T*构建完成 ({len(self.T_star.edges())} 条边)")
        
        # 步骤3: 识别非树边
        self.get_non_tree_edges()
        print(f"非树边数量: {len(self.non_tree_edges)}")
        
        # 步骤4: 枚举有效环
        self.find_valid_structures()
        
        # 步骤5: 生成候选解
        self.generate_candidates()
        print(f"生成候选解: {len(self.candidates)} 个")
        
        # 步骤6: 找到最优ATSP解
        tour_edges, cost = self.find_optimal_tour()
        print("tour s;fjkasjf;lajfkajasljfa")
        print(type(tour_edges))
        
        if tour_edges:
            print(f"🎯 找到ATSP最优解: 成本 = {cost:.4f}")
            self.visualize_tour(tour_edges, cost)
            return tour_edges, cost
        else:
            print("未找到Hamilton环解")
            return None, float('inf')


def generate_random_instance(n, dist_type='exp'):
    """生成随机ATSP实例"""
    if dist_type == 'exp':
        # 指数分布
        C = np.random.exponential(scale=1.0, size=(n, n))
    elif dist_type == 'uniform':
        # 均匀分布
        C = np.random.uniform(0, 1, size=(n, n))
    else:
        # 自定义分布 P(C≤x)=x+O(x²)
        C = np.zeros((n, n))
        for i in range(n):
            for j in range(n):
                u = np.random.uniform(0, 1)
                C[i][j] = u - 0.5 * u**2
    
    # 确保对角线不特殊处理
    np.fill_diagonal(C, np.random.uniform(0, 1, n))
    return C


def verify_solution(C, tour):
    """验证解的正确性"""
    cost = 0
    visited = set()
    current = 0
    
    for _ in range(len(C)):
        next_node = tour[current]
        cost += C[current][next_node]
        visited.add(next_node)
        current = next_node
    
    return len(visited) == len(C), cost


# ===================== 主程序 =====================
if __name__ == "__main__":
    np.random.seed(42)
    
    print("="*50)
    print("Bell & Frieze (2024) ATSP 算法实现")
    print("="*50)
    
    # 生成随机实例 (n=5)
    n = 5
    print(f"\n生成 {n} 城市随机 ATSP 实例...")
    C = generate_random_instance(n, dist_type='exp')
    for i in range(n):
        C[i][i] = 100
    print("成本矩阵:")
    print(np.round(C, 3))
    
    # 创建求解器
    solver = BellFriezeATSP(C)
    
    # 求解ATSP
    print("\n开始求解...")
    tour_edges, cost = solver.solve()
    
    if tour_edges:
        # 将边集转换为路径
        graph = {i: j for (i, j) in tour_edges}
        tour_path = [0]
        current = 0
        for _ in range(n-1):
            current = graph[current]
            tour_path.append(current)
        
        # 验证解
        valid, verified_cost = verify_solution(C, graph)
        print(f"最优路径: {tour_path}")
        print(f"验证: {'成功' if valid else '失败'}, 成本={verified_cost:.4f}")
        
        # 显示所有可视化
        solver.show_visualizations()
```
