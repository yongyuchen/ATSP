```python
# 基于成本矩阵的 TSP 求解程序
from pyscipopt import Model, Expr
import numpy as np
import matplotlib.pyplot as plt
from os import path

def add_mtz_subtour_constraints(n, model, x):
    """MTZ 子路径消除约束（精简版）"""
    # 创建辅助变量
    u = [model.addVar(lb=0, vtype="C", name=f"u_{i}") for i in range(n)]
    
    # 添加 MTZ 约束
    for i in range(1, n):
        for j in range(1, n):
            model.addCons(u[i] - u[j] + (n - 1) * x[i][j] <= n - 2)
        model.addCons(u[i] >= 1)
        model.addCons(u[i] <= n - 1)
    
    # 固定起点约束
    model.addCons(u[0] == 0)

def solve_tsp_from_matrix(cost_matrix, node_coords=None, show_plot=True):
    """
    基于成本矩阵的 TSP 求解函数
    
    cost_matrix: 成本矩阵 (n x n)
    node_coords: 节点坐标列表 (可选)
    show_plot: 是否显示图形结果 (需要坐标)
    """
    # 1. 获取问题维度
    n = len(cost_matrix)
    print(f"TSP 问题: {n}节点")
    
    # 2. 创建模型
    model = Model("tsp_matrix")
    
    # 3. 创建变量
    x = [[model.addVar(lb=0, ub=1, vtype="B", name=f"x_{i}_{j}") 
          for j in range(n)]
          for i in range(n)]
    
    # 4. 设置目标函数
    obj = Expr()
    for i in range(n):
        for j in range(n):
            if i != j:
                obj += cost_matrix[i][j] * x[i][j]
    model.setObjective(obj, "minimize")
    
    # 5. 添加基本约束
    for i in range(n):
        incoming = sum(x[j][i] for j in range(n))
        outgoing = sum(x[i][j] for j in range(n))
        model.addCons(incoming == 1, name=f"in_{i}")
        model.addCons(outgoing == 1, name=f"out_{i}")
        model.addCons(x[i][i] == 0, name=f"self_{i}")
    
    # 6. 添加 MTZ 约束
    add_mtz_subtour_constraints(n, model, x)
    
    # 7. 求解模型
    print("求解中...")
    model.hideOutput()  # 隐藏求解器详细输出
    model.optimize()
    
    # 8. 显示结果
    if model.getNSols() > 0:
        sol = model.getBestSol()
        objval = model.getObjVal(sol)
        
        # 提取解
        xval = [[model.getSolVal(sol, x[i][j]) 
                 for j in range(n)]
                 for i in range(n)]
        
        # 获取路径
        tour = []
        current = 0
        while len(tour) < n:
            tour.append(current)
            for j in range(n):
                if xval[current][j] > 0.5:
                    current = j
                    break
        tour.append(tour[0])
        
        # 计算实际成本
        actual_cost = 0
        for i in range(len(tour)-1):
            actual_cost += cost_matrix[tour[i]][tour[i+1]]
        
        # 显示结果
        print(f"\n最优成本: {actual_cost:.2f}")
        print(f"求解时间: {model.getTotalTime():.1f}秒")
        print(f"路径: {' → '.join(map(str, tour))}")
        return actual_cost
        
        # 显示图形 (如果有坐标)
        if show_plot and node_coords is not None and len(node_coords) == n:
            try:
                # 提取坐标
                coords = node_coords
                
                # 绘制图形
                plt.figure(figsize=(10, 8))
                
                # 绘制节点
                x_vals = [coords[node][0] for node in tour]
                y_vals = [coords[node][1] for node in tour]
                plt.plot(x_vals, y_vals, 'b-', marker='o', markersize=4)
                
                # 标记起点和终点
                plt.plot(coords[tour[0]][0], coords[tour[0]][1], 'ro', markersize=8)
                plt.plot(coords[tour[-2]][0], coords[tour[-2]][1], 'go', markersize=8)
                
                # 添加标签
                for i, coord in enumerate(coords):
                    plt.text(coord[0], coord[1], str(i), fontsize=8, ha='center', va='center')
                
                plt.title(f'TSP 路径 - 成本: {actual_cost:.2f}')
                plt.xlabel('X坐标')
                plt.ylabel('Y坐标')
                plt.grid(True, linestyle='--', alpha=0.3)
                plt.show()
            except Exception as e:
                print(f"图形显示错误: {str(e)}")
        elif show_plot and node_coords is None:
            print("未提供节点坐标，无法显示图形")
    else:
        print("未找到可行解")
```

```python
import numpy as np
from scipy.optimize import linear_sum_assignment
import math
from collections import deque

def karp_patching_atsp(distance_matrix, verbose=False):
    """
    完整的Karp修补算法实现，包含循环合并机制
    参数:
        distance_matrix (numpy.ndarray): n x n 非对称距离矩阵
        verbose (bool): 是否打印详细过程
    返回:
        tuple: (回路路径, 回路总长度)
    """
    n = distance_matrix.shape[0]
    
    # 1. 解决分配问题
    if verbose: print("步骤1: 求解分配问题...")
    row_ind, col_ind = linear_sum_assignment(distance_matrix)
    perm = np.zeros(n, dtype=int)
    perm[row_ind] = col_ind
    
    # 2. 分解置换为循环
    def decompose_cycles(permutation):
        """分解置换为不相交的循环"""
        visited = [False] * n
        cycles = []
        for i in range(n):
            if not visited[i]:
                cycle = []
                current = i
                while not visited[current]:
                    visited[current] = True
                    cycle.append(current)
                    current = permutation[current]
                cycles.append(cycle)
        return cycles
    
    # 3. 循环合并预处理
    #print("我是perm")
    #print(perm)
    cycles = decompose_cycles(perm)
    k = len(cycles)
    
    if verbose: 
        print(f"初始循环分解: 共{k}个循环")
        for i, cycle in enumerate(cycles):
            print(f"循环{i+1} (长度{len(cycle)}): {cycle}")
    
    # 4. 确保基础循环长度足够
    while True:
        base_cycle = max(cycles, key=len)
        k = len(cycles)
        
        # 检查是否满足条件
        if len(base_cycle) >= k - 1 or k == 1:
            if verbose: print(f"基础循环长度{len(base_cycle)}满足条件(k={k})")
            break
        
        if verbose: print(f"基础循环太短({len(base_cycle)} < {k-1}), 执行循环合并...")
        
        # 5. 寻找最小代价合并点
        min_delta = float('inf')
        best_pair = None
        best_cycles = None
        
        # 启发式优化：限制大循环的候选点
        MAX_CANDIDATES = min(50, max(10, int(math.sqrt(n))))
        
        for i in range(len(cycles)):
            for j in range(i+1, len(cycles)):
                cycle1 = cycles[i]
                cycle2 = cycles[j]
                
                # 采样大循环
                candidates1 = cycle1 if len(cycle1) <= MAX_CANDIDATES else np.random.choice(cycle1, MAX_CANDIDATES, replace=False)
                candidates2 = cycle2 if len(cycle2) <= MAX_CANDIDATES else np.random.choice(cycle2, MAX_CANDIDATES, replace=False)
                
                for a in candidates1:
                    for b in candidates2:
                        # 计算Δ值 (论文公式)
                        delta = (distance_matrix[a, perm[b]] 
                                 + distance_matrix[b, perm[a]] 
                                 - distance_matrix[a, perm[a]] 
                                 - distance_matrix[b, perm[b]])
                        
                        if delta < min_delta:
                            min_delta = delta
                            best_pair = (a, b)
                            best_cycles = (i, j)
        
        # 6. 执行最优合并
        if best_pair:
            a, b = best_pair
            if verbose: print(f"合并循环{best_cycles[0]+1}和{best_cycles[1]+1}, 使用点{a}-{b}, Δ={min_delta:.2f}")
            
            # 执行修补操作（交换后继节点）
            perm[a], perm[b] = perm[b], perm[a]
            
            # 重新分解循环
            cycles = decompose_cycles(perm)
            k = len(cycles)
        else:
            raise RuntimeError("无法找到有效的合并点")
    
    # 7. 构建修补矩阵
    base_cycle = max(cycles, key=len)
    base_cycle_set = set(base_cycle)
    other_cycles = [c for c in cycles if c != base_cycle]
    k = len(cycles)
    
    if verbose: 
        print(f"最终基础循环: 长度{len(base_cycle)}")
        print(f"需要合并的循环数: {len(other_cycles)}")
    
    # 8. 创建(k-1)×|base_cycle|修补矩阵
    A = np.full((len(other_cycles), len(base_cycle)), np.inf)
    L = np.full((len(other_cycles), len(base_cycle)), -1, dtype=int)  # 记录最小Δ对应的元素
    
    for i, cycle in enumerate(other_cycles):
        for j_idx, j in enumerate(base_cycle):
            min_delta = np.inf
            best_l = -1
            
            for l in cycle:
                # 计算Δ值
                delta = (distance_matrix[l, perm[j]] 
                         + distance_matrix[j, perm[l]] 
                         - distance_matrix[l, perm[l]] 
                         - distance_matrix[j, perm[j]])
                
                if delta < min_delta:
                    min_delta = delta
                    best_l = l
            
            A[i, j_idx] = min_delta
            L[i, j_idx] = best_l
    
    # 9. 解决子分配问题
    row_ind, col_ind = linear_sum_assignment(A)
    
    # 10. 执行修补操作
    for i, j_idx in zip(row_ind, col_ind):
        l = L[i, j_idx]
        j = base_cycle[j_idx]
        
        if verbose: print(f"执行修补: 循环{i+1}的点{l} ↔ 基础循环的点{j} (Δ={A[i, j_idx]:.2f})")
        
        # 交换l和j的后继
        perm[l], perm[j] = perm[j], perm[l]
    
    # 11. 从最终置换构建回路
    # 使用BFS确保正确路径顺序
    visited = [False] * n
    tour = []
    queue = deque([0])
    
    while queue:
        current = queue.popleft()
        if visited[current]:
            continue
            
        visited[current] = True
        tour.append(current)
        next_node = perm[current]
        
        if not visited[next_node]:
            queue.append(next_node)
    
    # 添加起点完成回路
    tour.append(tour[0])
    
    # 12. 计算回路长度
    cost = 0
    for i in range(len(tour)-1):
        cost += distance_matrix[tour[i], tour[i+1]]
    
    if verbose: print(f"最终回路: {tour}, 总长度: {cost:.2f}")
    
    return tour, cost
```

```python
import numpy as np
from scipy.optimize import linear_sum_assignment
%matplotlib inline
import matplotlib.pyplot as plt
import networkx as nx
import math
from itertools import combinations
import matplotlib.colors as mcolors
from scipy.sparse.csgraph import connected_components
from scipy.sparse import csr_matrix
from pyscipopt import Model  # 添加SCIP求解器
from concurrent.futures import ThreadPoolExecutor
import logging
# 使用更高效的并行处理框架
from joblib import Parallel, delayed
# 使用近似聚类替代精确社区检测
from sklearn.cluster import DBSCAN
from matplotlib.lines import Line2D # 确保导入 Line2D 用于图例
import random  # 导入 random 库
import matplotlib
from collections import Counter
import re

# 配置日志
logging.basicConfig(level=logging.INFO, 
                    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

class BellFriezeATSP:
    __slots__ = ['C', 'n', 'ap_edges', 'ap_cost', 'uu', 'vv', 'non_tree_edges', 'candidates', 'debug', 'logger', 'T_star', 'valid_cycles', 
                'figures', 'k_max', 'base_graph', 'base_graph_edges', 'T_edges', 'bipartite_edges', 'clusters_result', 'num_clusters']  # 仅声明必需字段
    def __init__(self, cost_matrix, debug=False):
        # 输入验证
        if not isinstance(cost_matrix, np.ndarray):
            raise TypeError("成本矩阵必须为NumPy数组")
        if cost_matrix.shape[0] != cost_matrix.shape[1]:
            raise ValueError("成本矩阵必须为方阵")
        if np.any(cost_matrix < 0):
            raise ValueError("成本不能为负值")
            
        #super().__init__(*args, **kwargs)
        self.C = cost_matrix
        self.n = len(cost_matrix)
        #self.spatial_threshold = 0.3 * self.n
        self.debug = debug
        self.logger = logging.getLogger('ATSP-Solver')
        self.logger.info(f"初始化ATSP求解器，城市数: {self.n}")
        
        # 算法数据结构
        self.ap_edges = None
        self.ap_cost = None
        # 使用边列表代替完整的networkx图对象
        self.T_edges = set()  # 存储生成树边(i, j)而不是完整的图对象
        #self.T_star = None
        self.non_tree_edges = None
        self.valid_cycles = None
        self.bipartite_edges = None
        self.candidates = {}
        #self.compute_new_edges = {}
        self.figures = []  # store images
        
        # 对偶变量
        self.uu = np.zeros(self.n)
        self.vv = np.zeros(self.n)
        
        # 优化参数：限制非树边组合大小
        self.k_max = min(5, max(1, math.ceil(math.log(self.n)**2.5)))
        self.logger.debug(f"设置k_max={self.k_max}")
        
    def solve_ap(self):
        """使用SCIP求解指派问题并获取对偶变量"""
        n = self.n
        model = Model("AssignmentProblem")
        
        # 创建变量
        x = {}
        for i in range(n):
            for j in range(n):
                x[i, j] = model.addVar(vtype="C", lb=0, ub=1, name=f"x_{i}_{j}")
        
        # 设置目标函数：最小化总成本
        model.setObjective(sum(self.C[i, j] * x[i, j] for i in range(n) for j in range(n)), "minimize")
        
        # 添加行约束：每行和为1
        row_constraints = []
        for i in range(n):
            cons = model.addCons(sum(x[i, j] for j in range(n)) == 1, name=f"row_{i}")
            row_constraints.append(cons)
        
        # 添加列约束：每列和为1
        col_constraints = []
        for j in range(n):
            cons = model.addCons(sum(x[i, j] for i in range(n)) == 1, name=f"col_{j}")
            col_constraints.append(cons)
        
        # 求解模型
        model.optimize()
        
        # 检查求解状态
        if model.getStatus() != "optimal":
            raise RuntimeError("SCIP failed to find optimal solution")
        
        # 保存最优匹配边
        self.ap_edges = []
        for i in range(n):
            for j in range(n):
                if model.getVal(x[i, j]) > 0.5:  # 大于0.5视为选中
                    self.ap_edges.append((i, j))

        # 计算并存储AP成本
        self.ap_cost = model.getObjVal()
        print(f"AP最优成本: {self.ap_cost:.4f}")
        
        print("ap edges")
        print(self.ap_edges)
        # 提取对偶变量
        self.uu = np.zeros(n)
        self.vv = np.zeros(n)
        
        for i in range(n):
            self.uu[i] = model.getDualsolLinear(row_constraints[i])
        
        for j in range(n):
            self.vv[j] = model.getDualsolLinear(col_constraints[j])
            
        # 添加对偶验证
        dual_sum = sum(self.uu) + sum(self.vv)
        if abs(self.ap_cost - dual_sum) > 1e-5:
            self.logger.warning(f"对偶间隙: {abs(self.ap_cost - dual_sum):.6f}")

        # Visualization AP solution
        self.visualize_matching(self.ap_edges, "AP optimal solutions", f"cost: {self.ap_cost:.4f}")
        self.visualize_matching2(self.ap_edges, "AP optimal solutions", f"cost: {self.ap_cost:.4f}")
        self.visualize_matching3(self.ap_edges, "AP optimal solutions", f"cost: {self.ap_cost:.4f}")
    
    def build_spanning_tree(self):
        """并行构建生成树"""
        tol = 1e-5
        self.logger.info("开始构建生成树...")

        
        # 并行识别紧边
        with ThreadPoolExecutor() as executor:
            results = list(executor.map(self._find_row_tight_edges, range(self.n)))
        tight_edges = [item for sublist in results for item in sublist]
        
        
        # 构建基础图（后续复用）
        self.base_graph = nx.Graph()
        parent = list(range(2 * self.n))
        T_edges = set(self.ap_edges)
        
        # ... 并查集操作不变 ...
        
        n = self.n
        
        # 步骤3: 合并匹配边
        for i, j in self.ap_edges:
            self.union(parent, i, j + n)
        
        # 步骤4: 添加非匹配紧边连接分量
        for (i, j) in tight_edges:
            if (i, j) in self.ap_edges:  # 跳过匹配边
                continue
                
            a_node = i
            b_node = j + n
            if self.find(parent, a_node) != self.find(parent, b_node):
                self.union(parent, a_node, b_node)
                T_edges.add((i, j))
                if len(T_edges) == 2 * n - 1:  # 已达2n-1条边
                    break

        # 添加到基础图
        #for (i, j) in T_edges:
            #self.base_graph.add_edge(i, j+self.n)
        #self.T_star = self.base_graph.copy()
        for (i, j) in T_edges:
        # 存储为二分图边 (i, j+n)
            self.T_edges.add((i, j + self.n))  # 关键修改
        self.bipartite_edges = list(self.T_edges)  # 用于后续环检测
        #for (i, j) in T_edges:
            #self.T_edges.add((i, j))  # 仅存储边元组
        #self.T_star = self.T_edges.copy()
        self.logger.info(f"生成树构建完成，边数: {len(T_edges)}")

        # 释放基础图内存
        #del self.base_graph
        #self.base_graph_edges = list(self.T_star.edges())  # 只存储边元组
        
        # 构建生成树图，使用整数节点
        #self.T_star = nx.Graph()
        #for (i, j) in T_edges:
            # 添加边：A部节点i，B部节点j+n
            #self.T_star.add_edge(i, j+self.n, weight=self.C[i, j])
    
        #return list(T_edges), M_star, u, v
        #self.T_star = T_edges
        #A = nx.to_numpy_array(self.T_star)
        #print("YYC spaning tree hrefdsjflsjakl'fjasfjaslfkafafa\n")
        #print(T_edges)
        #print(A)
        
        
        # Visualization the spaning tree
        self.visualize_spanning_tree(self.T_edges)
        #self.visualize_spanning_tree2(self.T_edges)
        self.visualize_matching4(self.ap_edges, "spanning tree", f"cost: {self.ap_cost:.4f}", tight_edges)
        #return T_edges

    # 并查集工具函数"
    @staticmethod
    def find(parent, x):
        """查找根节点（带路径压缩）"""
        if parent[x] != x:
            parent[x] = BellFriezeATSP.find(parent, parent[x])
        return parent[x]
    
    @staticmethod
    def union(parent, x, y):
        """合并两个集合"""
        root_x = BellFriezeATSP.find(parent, x)
        root_y = BellFriezeATSP.find(parent, y)
        if root_x != root_y:
            parent[root_y] = root_x
            
    def _find_row_tight_edges(self, i):
        """辅助函数：并行处理单行"""
        row = []
        for j in range(self.n):
            diff = abs(self.C[i, j] - (self.uu[i] + self.vv[j]))
            if diff < 1e-5:
                row.append((i, j))
        return row
    
    def get_non_tree_edges(self):
        self.non_tree_edges = []
        n = self.n
        tol = 1e-5
        
        # 获取树边集合（使用整数标识）
        tree_edges = self.T_edges
        
        for i in range(n):
            for j in range(n):
                # 统一使用整数标识
                is_tree_edge = (i, j) in tree_edges
                is_tight = abs(self.C[i, j] - (self.uu[i] + self.vv[j])) < tol
                
                if not is_tree_edge or not is_tight:
                    self.non_tree_edges.append((i, j))
        
        #print("here is non tree edges")
        
        # Visualize all non-basic edges
        #self.visualize_non_tree_edges()
        self.visualize_matching5(self.ap_edges, "AP optimal solutions", f"cost: {self.ap_cost:.4f}", self.non_tree_edges)
        #return self.non_tree_edges

    def process_cluster_wrapper(self, args):
        return process_cluster(*args)
    #def process_cluster_wrapper(cluster):
        #return process_cluster(cluster, self.T_edges, self.n)
        
    def cluster_edges(self, edges):
        """
        使用简单的 2x2 网格进行空间聚类。
        
        这种方法非常快，并且保证了结果是 4 个簇 (除非某些网格为空)，
        完美匹配 n_jobs=4 的并行处理需求。
        """
        if not edges:
            return []
            
        # 将空间在 i 和 j 维度上都一分为二
        mid_point = self.n / 2.0
        
        # [Top-Left, Top-Right, Bottom-Left, Bottom-Right]
        clusters = [[], [], [], []]
        
        for edge in edges:
            i, j = edge
            
            if i < mid_point and j < mid_point:
                clusters[0].append(edge)  # 簇 0: 左上
            elif i < mid_point and j >= mid_point:
                clusters[1].append(edge)  # 簇 1: 右上
            elif i >= mid_point and j < mid_point:
                clusters[2].append(edge)  # 簇 2: 左下
            else:
                clusters[3].append(edge)  # 簇 3: 右下
        
        # 过滤掉空簇
        non_empty_clusters = [c for c in clusters if c]
        
        msg = f"网格聚类: 已将边分为 {len(non_empty_clusters)} 个非空区域 (共4个网格)"
        if hasattr(self, 'logger'):
            self.logger.info(msg)
        else:
            print(msg)

        return non_empty_clusters

    # 优化：基于几何位置剪枝
    def is_spatial_close(self, edge1, edge2, threshold):
        """检查两条边是否空间接近"""
        #threshold = 0.3 * self.n
        #threshold = self.spatial_threshold
        i1, j1 = edge1
        i2, j2 = edge2
        return abs(i1-i2) + abs(j1-j2) < threshold
        

    def find_valid_structures(self):
        """使用joblib并行处理"""
        from joblib import Parallel, delayed
        
        # 1. 调用聚类方法，并将结果存储在 self.clusters_result 中
        #    (我们不用 self.cluster_edges 命名，以避免与同名方法冲突)
        self.clusters_result = self.cluster_edges(self.non_tree_edges)
        
        # 2. 现在你可以立即知道簇的数量
        self.num_clusters = len(self.clusters_result)
        
        # 3. 将这个信息打印出来
        msg = f"已将 {len(self.non_tree_edges)} 条边分为 {self.num_clusters} 个簇，准备并行处理..."
        if hasattr(self, 'logger'):
            self.logger.info(msg)
        else:
            print(msg)
            
        self.valid_cycles = []
        
        # 4. 并行处理
        with Parallel(n_jobs=4, prefer="threads") as parallel:
            # 使用 self.clusters_result 进行迭代
            results = parallel(
                delayed(process_cluster)(cluster, self.bipartite_edges, self.n)
                for cluster in self.clusters_result 
            )
            for res in results:
                self.valid_cycles.extend(res)
        
        self.logger.info(f"找到有效环: {len(self.valid_cycles)}个")
        total_cycles_found = len(self.valid_cycles)

        if total_cycles_found == 0:
            self.logger.info("未找到任何有效环。")
        else:
            # 1. 使用 Counter 按环的长度(边数)进行分类
            cycle_length_counter = Counter(len(cycle) for cycle in self.valid_cycles)
            
            # 2. 打印一个清晰的总结报告
            self.logger.info(f"找到有效环总数: {total_cycles_found} 个，分类如下:")
            
            # 按边数(key)排序后打印
            for num_edges, count in sorted(cycle_length_counter.items()):
                # 因为它们是交替环, k (非树边数) = 边数 / 2
                k_value = num_edges / 2
                self.logger.info(f"    - {num_edges} 边环 (k={k_value:.0f}): {count} 个")
        self.visualize_matching_with_clusters(self.ap_edges, "AP optimal solutions", f"cost: {self.ap_cost:.4f}", self.non_tree_edges, self.clusters_result)
        self.visualize_valid_cycles(self.ap_edges, "valid cycles", subtitle=f"total number {len(self.valid_cycles)}")
        self.visualize_individual_cycles(self.ap_edges, num_to_plot=5, title_prefix="found valid cycles")
        
    def compute_new_edges(self, cycle_edges):
        """
        计算通过环路替换后的新边集
        :param cycle_edges: 有效环的边集合
        :return: 新边集 (AP解与环边的对称差)
        """
        new_edges = set(self.ap_edges)  # 复制AP解边集
        for edge in cycle_edges:
            if edge in new_edges:
                new_edges.remove(edge)  # 移除已存在的边
            else:
                new_edges.add(edge)     # 添加新边
        return new_edges
    
    def generate_candidates(self, max_tours_to_plot=5, max_failed_to_plot=5):
        """
        【已修改】
        生成候选解，并绘制“成功”(哈密顿环)和“失败”(非哈密顿环)的样本。
        
        :param max_tours_to_plot: 最多绘制多少个“成功”的哈密顿环。
        :param max_failed_to_plot: 最多绘制多少个“失败”的候选解。
        """
        best_cost = float('inf')
        best_tour = None
        
        # AP解 (您的原始代码)
        # if self.is_hamiltonian(self.ap_edges):
        # ... (省略)
        
        # 用于计数的变量
        found_tours_count = 0 
        found_failed_count = 0  # <-- 新增的计数器
        
        # 有效环解
        for cycle_edges in self.valid_cycles:
            new_edges = self.compute_new_edges(cycle_edges)
            
            # 【【【*** 核心修改点 ***】】】
            if self.is_hamiltonian(new_edges):
                
                # 1. 计算成本
                cost = sum(self.C[i][j] for (i, j) in new_edges)
                
                # 2. 检查是否是新的最优解
                is_new_best = (cost < best_cost)
                
                # 3. 绘图 (有条件地)
                
                # A) 如果它是新的最优解，*总是*绘制它
                if is_new_best:
                    best_cost = cost
                    best_tour = new_edges
                    
                    self.logger.info(f"*** 找到新的最优哈密顿环！ 成本: {best_cost:.4f} ***")
                    # 调用新函数名，并使用 绿色
                    self.visualize_loops( 
                        best_tour, 
                        title=f"*** new optimal hamiltonian cycles ***",
                        subtitle=f"cost (Cost): {best_cost:.4f}",
                        node_color='green' # 绿色代表成功
                    )
                
                # B) 否则，如果它只是一个“普通”的环...
                elif found_tours_count < max_tours_to_plot:
                    self.logger.info(f"找到一个哈密顿环 (成本: {cost:.4f})，正在绘制...")
                    # 调用新函数名，并使用 蓝色
                    self.visualize_loops(
                        new_edges, 
                        title=f"found hamiltonian cycles (candidate {found_tours_count + 1})",
                        subtitle=f"cost (Cost): {cost:.4f}",
                        node_color='blue' # 蓝色代表“合格但非最优”
                    )
                    found_tours_count += 1
            
            # 【【【*** 新增的 ELSE 块 ***】】】
            else:
                # 这是一个“失败”的候选解 (非哈密顿环)
                # 我们只画前 max_failed_to_plot 个
                if found_failed_count < max_failed_to_plot:
                    
                    self.logger.info(f"找到一个“失败”的候选解 (非哈密顿环)，正在绘制样本 {found_failed_count + 1}...")
                    
                    # 调用新函数名，并使用 红色
                    self.visualize_loops(
                        new_edges,
                        title=f"failed candidate (sample {found_failed_count + 1})",
                        subtitle="(Non-Hamiltonian)",
                        node_color='red' # 红色代表失败
                    )
                    found_failed_count += 1
                
                # (如果既不是哈密顿环，也超出了绘图限制，就跳过绘图)
                
        # 循环结束后的总结
        if best_tour is None:
             self.logger.warning("未能从任何有效环中生成哈密顿环。")
        else:
             self.logger.info(f"搜索完成。已找到最优哈密顿环，成本: {best_cost:.4f}")

        return best_tour, best_cost
    
    
    def find_optimal_tour(self):
        """找到最优ATSP解"""
        optimal_cost = float('inf')
        optimal_tour = None
        print("optimal tour YYC\n")
        print(optimal_tour)
        
        for edges, cost in self.candidates.items():
            if self.is_hamiltonian(edges) and cost < optimal_cost:
                optimal_cost = cost
                optimal_tour = edges
        
        return optimal_tour, optimal_cost
    
    def is_hamiltonian(self, edges):
        """高效检查Hamilton回路"""
        if len(edges) != self.n:  # 边数必须等于节点数
            return False
        
        graph = {}
        for i, j in edges:
            if i in graph:
                return False  # 出度>1
            graph[i] = j
        
        visited = set()
        node = 0
        for _ in range(self.n):
            visited.add(node)
            node = graph.get(node)
            if node is None:
                return False
        
        return len(visited) == self.n
    
    def visualize_matching(self, edges, title, subtitle=""):
        """可视化匹配（有向图版本）"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))  # 添加所有节点
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:  # 避免自环
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', style='dashed', alpha=0.3, arrowstyle='-|>', arrowsize=10)
    
        # 添加匹配边（红色实线）
        for i, j in edges:
            G.add_edge(i, j, color='red', width=2.0, arrowstyle='-|>', arrowsize=15)
    
        # 使用环形布局
        pos = nx.circular_layout(G)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(
            G, pos, edgelist=all_edges,
            edge_color='lightgray', width=0.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3
        )
    
        # 绘制匹配边（红色实线）
        match_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'red']
        nx.draw_networkx_edges(
            G, pos, edgelist=match_edges,
            edge_color='red', width=2.0, 
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 添加标题
        plt.title(f"{title}\n{subtitle}", fontsize=14)
        plt.axis('off')
        plt.tight_layout()

    def visualize_matching2(self, edges, title, subtitle=""):
        """可视化匹配（二部图版本）"""
        fig = plt.figure(figsize=(6, 4))
        self.figures.append(fig)
        
        # 创建无向图
        G = nx.Graph()
        
        # 创建左右两个节点集
        left_nodes = [f"L{i}" for i in range(self.n)]
        right_nodes = [f"R{i}" for i in range(self.n)]
        G.add_nodes_from(left_nodes, bipartite=0)
        G.add_nodes_from(right_nodes, bipartite=1)
        
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                G.add_edge(f"L{i}", f"R{j}", weight=self.C[i][j], 
                          color='lightgray', style='dashed', width=0.5)
        
        # 标记匹配边
        for i, j in edges:
            G[f"L{i}"][f"R{j}"]["color"] = "red"
            G[f"L{i}"][f"R{j}"]["width"] = 2.5
        
        # 二部图布局
        pos = {}
        # 左侧节点
        for i, node in enumerate(left_nodes):
            pos[node] = (0, (self.n - 1) / 2 - i)
        # 右侧节点
        for i, node in enumerate(right_nodes):
            pos[node] = (2, (self.n - 1) / 2 - i)
        
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, nodelist=left_nodes, 
                              node_size=500, node_color='lightblue', 
                              node_shape='o', label='Left set')
        nx.draw_networkx_nodes(G, pos, nodelist=right_nodes, 
                              node_size=500, node_color='lightcoral', 
                              node_shape='s', label='Right set')
        
        # 绘制标签（去掉L和R前缀）
        labels = {f"L{i}": str(i) for i in range(self.n)}
        labels.update({f"R{i}": str(i) for i in range(self.n)})
        nx.draw_networkx_labels(G, pos, labels=labels, 
                               font_size=10, font_weight='bold')
        
        # 绘制所有可能的边（灰色虚线）
        all_edges = [(u, v) for u, v in G.edges() 
                     if G[u][v].get('color') == 'lightgray']
        nx.draw_networkx_edges(G, pos, edgelist=all_edges,
                              edge_color='lightgray', width=0.5, 
                              style='dashed', alpha=0.3)
        
        # 绘制匹配边（红色实线）
        match_edges = [(u, v) for u, v in G.edges() 
                       if G[u][v].get('color') == 'red']
        nx.draw_networkx_edges(G, pos, edgelist=match_edges,
                              edge_color='red', width=2.5, alpha=0.8)
        
        # 添加标题和图例
        plt.title(f"{title}\n{subtitle}", fontsize=14)
        plt.legend(loc='upper right', fontsize=9)
        plt.axis('off')
        plt.tight_layout()

    def visualize_matching3(self, edges, title, subtitle=""):
        """可视化匹配（有向图多环形式）"""
        fig = plt.figure(figsize=(8, 6))
        self.figures.append(fig)
        
        # 构建匹配关系
        matching = {}
        for i, j in edges:
            matching[i] = j
        
        # 找出所有的环
        visited = set()
        cycles = []
        
        for start in range(self.n):
            if start in visited:
                continue
            
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            
            # 检查是否形成环
            if len(cycle) > 1 and current == start:
                cycles.append(cycle)
            elif len(cycle) == 1 and start in matching and matching[start] == start:
                cycles.append(cycle)  # 自环
        
        # 如果没有环，单独处理每条边
        if not cycles:
            for i, j in edges:
                cycles.append([i, j] if i != j else [i])
        
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 为每个环分配颜色
        colors = plt.cm.tab10(range(len(cycles)))
        
        # 计算每个环的位置
        num_cycles = len(cycles)
        cols = int(np.ceil(np.sqrt(num_cycles)))
        rows = int(np.ceil(num_cycles / cols))
        
        all_pos = {}
        
        for idx, cycle in enumerate(cycles):
            # 计算环的中心位置
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            
            # 为环中的节点创建圆形布局
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3)
            
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (
                    center_x + radius * np.cos(angle),
                    center_y + radius * np.sin(angle)
                )
        
        # 绘制每个环
        for idx, cycle in enumerate(cycles):
            color = colors[idx]
            
            # 绘制环中的节点
            cycle_nodes = cycle
            nx.draw_networkx_nodes(
                G, all_pos, nodelist=cycle_nodes,
                node_size=600, node_color=[color], 
                edgecolors='black', linewidths=2
            )
            
            # 绘制环中的有向边
            cycle_edges = []
            for i in range(len(cycle)):
                u = cycle[i]
                v = cycle[(i + 1) % len(cycle)]
                if u in matching and matching[u] == v:
                    cycle_edges.append((u, v))
            
            # 绘制有向匹配边
            nx.draw_networkx_edges(
                G, all_pos, edgelist=cycle_edges,
                edge_color=[color], width=3.0, alpha=0.8,
                arrows=True, arrowstyle='-|>', arrowsize=20,
                connectionstyle='arc3,rad=0.0', 
                node_size=600
            )
            
            # 如果只有一个节点，绘制有向自环
            if len(cycle) == 1 and cycle[0] in matching and matching[cycle[0]] == cycle[0]:
                node = cycle[0]
                pos_node = all_pos[node]
                # 绘制自环圆圈
                circle = plt.Circle(
                    (pos_node[0] + 0.4, pos_node[1]), 0.25,
                    fill=False, color=color, linewidth=2.5
                )
                plt.gca().add_patch(circle)
                # 添加箭头表示方向
                arrow = plt.Arrow(
                    pos_node[0] + 0.4 + 0.25, pos_node[1],
                    0.15, 0, width=0.2, color=color
                )
                plt.gca().add_patch(arrow)
        
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels, 
                               font_size=11, font_weight='bold')
        
        # 添加标题
        title_text = f"{title}\n{subtitle}\nfind(s) {len(cycles)} cycles"
        plt.title(title_text, fontsize=14, pad=20)
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout()
    
    def visualize_spanning_tree(self, T):
        """可视化生成树（有向图版本）"""
        fig = plt.figure(figsize=(6, 5))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', style='dashed', alpha=0.3)
    
        # 添加生成树中的边（蓝色实线）
        tree_edges = set()
        for u, v in T:
            # 转换二分图边为有向图边
            if u < self.n and v >= self.n:  # A->B边
                i, j = u, v - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
            elif v < self.n and u >= self.n:  # B->A边（转换为反向）
                i, j = v, u - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
    
        # 添加AP解中的边（红色实线）
        ap_edges = set()
        for i, j in self.ap_edges:
            ap_edges.add((i, j))
            if (i, j) not in tree_edges:  # 如果不在树中，添加为红色
                G.add_edge(i, j, color='red', width=2.0)
    
        # 使用弹簧布局
        pos = nx.spring_layout(G, k=0.5, iterations=50)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(G, pos, edgelist=all_edges, edge_color='lightgray', width=0.5, style='dashed',
        arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3)
    
        # 绘制生成树边（蓝色实线）
        tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'blue']
        nx.draw_networkx_edges(
            G, pos, edgelist=tree_edges,
            edge_color='blue', width=1.5, 
            arrows=True, arrowstyle='-|>', arrowsize=12)
    
        # 绘制AP解边（红色实线）
        ap_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'red']
        nx.draw_networkx_edges(
            G, pos, edgelist=ap_edges,
            edge_color='red', width=2.0, 
            arrows=True, arrowstyle='-|>', arrowsize=15)
    
        # 添加标题和图例
        plt.title(f"Spaning tree T* (contains AP solutions)", fontsize=14)
    
        # 创建图例
        import matplotlib.patches as mpatches
        gray_patch = mpatches.Patch(color='lightgray', label='possible edges')
        blue_patch = mpatches.Patch(color='blue', label='tree edges')
        red_patch = mpatches.Patch(color='red', label='AP solutions edges')
        plt.legend(handles=[gray_patch, blue_patch, red_patch], loc='best')
    
        plt.axis('off')
        plt.tight_layout()

    def visualize_matching4(self, edges, title, subtitle="", tight_edges=None):
        """可视化匹配（有向图多环形式）
        
        Args:
            edges: 匹配边列表
            title: 标题
            subtitle: 副标题
            tight_edges: 紧边列表（可选），将用虚线显示
        """
        fig = plt.figure(figsize=(8, 6))
        self.figures.append(fig)
        
        # 构建匹配关系
        matching = {}
        for i, j in edges:
            matching[i] = j
        
        # 找出所有的环
        visited = set()
        cycles = []
        
        for start in range(self.n):
            if start in visited:
                continue
            
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            
            # 检查是否形成环
            if len(cycle) > 1 and current == start:
                cycles.append(cycle)
            elif len(cycle) == 1 and start in matching and matching[start] == start:
                cycles.append(cycle)  # 自环
        
        # 如果没有环，单独处理每条边
        if not cycles:
            for i, j in edges:
                cycles.append([i, j] if i != j else [i])
        
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 为每个环分配颜色
        colors = plt.cm.tab10(range(len(cycles)))
        
        # 计算每个环的位置
        num_cycles = len(cycles)
        cols = int(np.ceil(np.sqrt(num_cycles)))
        rows = int(np.ceil(num_cycles / cols))
        
        all_pos = {}
        
        for idx, cycle in enumerate(cycles):
            # 计算环的中心位置
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            
            # 为环中的节点创建圆形布局
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3)
            
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (
                    center_x + radius * np.cos(angle),
                    center_y + radius * np.sin(angle)
                )
        
        # 绘制每个环
        for idx, cycle in enumerate(cycles):
            color = colors[idx]
            
            # 绘制环中的节点
            cycle_nodes = cycle
            nx.draw_networkx_nodes(
                G, all_pos, nodelist=cycle_nodes,
                node_size=600, node_color=[color], 
                edgecolors='black', linewidths=2
            )
            
            # 绘制环中的有向边（实线）
            cycle_edges = []
            for i in range(len(cycle)):
                u = cycle[i]
                v = cycle[(i + 1) % len(cycle)]
                if u in matching and matching[u] == v:
                    cycle_edges.append((u, v))
            
            # 绘制有向匹配边（实线）
            nx.draw_networkx_edges(
                G, all_pos, edgelist=cycle_edges,
                edge_color=[color], width=3.0, alpha=0.8,
                arrows=True, arrowstyle='-|>', arrowsize=20,
                connectionstyle='arc3,rad=0.0', 
                node_size=600, style='solid'
            )
            
            # 绘制紧边（虚线）
            if tight_edges is not None:
                tight_edges_in_cycle = []
                tight_edges_set = set(edges)  # 匹配边集合
                
                for u, v in tight_edges:
                    # 只绘制在当前环中的紧边，且不是匹配边
                    if u in cycle and v in cycle and (u, v) not in tight_edges_set:
                        tight_edges_in_cycle.append((u, v))
                
                if tight_edges_in_cycle:
                    nx.draw_networkx_edges(
                        G, all_pos, edgelist=tight_edges_in_cycle,
                        edge_color='gray', width=1.5, alpha=0.5,
                        arrows=True, arrowstyle='-|>', arrowsize=15,
                        connectionstyle='arc3,rad=0.1',  # 稍微弯曲以区分
                        node_size=600, style='dashed'
                    )
            
            # 如果只有一个节点，绘制有向自环
            if len(cycle) == 1 and cycle[0] in matching and matching[cycle[0]] == cycle[0]:
                node = cycle[0]
                pos_node = all_pos[node]
                # 绘制自环圆圈
                circle = plt.Circle(
                    (pos_node[0] + 0.4, pos_node[1]), 0.25,
                    fill=False, color=color, linewidth=2.5
                )
                plt.gca().add_patch(circle)
                # 添加箭头表示方向
                arrow = plt.Arrow(
                    pos_node[0] + 0.4 + 0.25, pos_node[1],
                    0.15, 0, width=0.2, color=color
                )
                plt.gca().add_patch(arrow)
        
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels, 
                               font_size=11, font_weight='bold')
        
        # 添加图例
        from matplotlib.lines import Line2D
        legend_elements = [
            Line2D([0], [0], color='black', linewidth=3.0, label='matching edges'),
        ]
        if tight_edges is not None:
            legend_elements.append(
                Line2D([0], [0], color='gray', linewidth=1.5, 
                       linestyle='dashed', label='tight edges')
            )
        plt.legend(handles=legend_elements, loc='upper right', fontsize=10)
        
        # 添加标题
        tight_info = f", {len(tight_edges)} tight edges" if tight_edges else ""
        title_text = f"{title}\n{subtitle}\nfind(s) {len(cycles)} cycles{tight_info}"
        plt.title(title_text, fontsize=14, pad=20)
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout()
    
    def visualize_non_tree_edges(self):
        """可视化非树边（有向图版本）"""
        fig = plt.figure(figsize=(6, 5))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', 
                          style='dashed', alpha=0.3)
    
        # 添加树边（蓝色实线）
        tree_edges = set()
        for u, v in self.T_star.edges():
            # 转换二分图边为有向图边
            if u < self.n and v >= self.n:  # A->B边
                i, j = u, v - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
            elif v < self.n and u >= self.n:  # B->A边（转换为反向）
                i, j = v, u - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
    
        # 添加非树边（绿色虚线）
        for (i, j) in self.non_tree_edges:
            # 确保边不存在于树边中
            if (i, j) not in tree_edges:
                G.add_edge(i, j, color='green', width=2.0, style='dashed')
        
        # 使用弹簧布局
        pos = nx.spring_layout(G, k=0.5, iterations=50)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(
            G, pos, edgelist=all_edges,
            edge_color='lightgray', width=0.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3
        )
    
        # 绘制树边（蓝色实线）
        tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'blue']
        nx.draw_networkx_edges(
            G, pos, edgelist=tree_edges,
            edge_color='blue', width=1.5, 
            arrows=True, arrowstyle='-|>', arrowsize=12
        )
    
        # 绘制非树边（绿色虚线）
        non_tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'green']
        nx.draw_networkx_edges(
            G, pos, edgelist=non_tree_edges,
            edge_color='green', width=2.0, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 添加标题和图例
        plt.title("Spaning tree and non-tree edges", fontsize=14)
    
        # 创建图例
        import matplotlib.patches as mpatches
        gray_patch = mpatches.Patch(color='lightgray', label='possible edges')
        blue_patch = mpatches.Patch(color='blue', label='tree edges')
        green_patch = mpatches.Patch(color='green', label='non-tree edges')
        plt.legend(handles=[gray_patch, blue_patch, green_patch], loc='best')
    
        plt.axis('off')
        plt.tight_layout()

    def visualize_matching5(self, edges, title, subtitle="", non_tight_edges=None):
        """可视化匹配（有向图多环形式）"""
        fig = plt.figure(figsize=(8, 6))
        self.figures.append(fig)
        
        # 构建匹配关系
        matching = {}
        for i, j in edges:
            matching[i] = j
        
        # 找出所有的环
        visited = set()
        cycles = []
        
        for start in range(self.n):
            if start in visited:
                continue
            
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            
            # 检查是否形成环
            if len(cycle) > 1 and current == start:
                cycles.append(cycle)
            elif len(cycle) == 1 and start in matching and matching[start] == start:
                cycles.append(cycle)  # 自环
        
        # 如果没有环，单独处理每条边
        if not cycles:
            for i, j in edges:
                cycles.append([i, j] if i != j else [i])
        
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 为每个环分配颜色
        colors = plt.cm.tab10(range(len(cycles)))
        
        # 计算每个环的位置
        num_cycles = len(cycles)
        cols = int(np.ceil(np.sqrt(num_cycles)))
        rows = int(np.ceil(num_cycles / cols))
        
        all_pos = {}
        
        for idx, cycle in enumerate(cycles):
            # 计算环的中心位置
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            
            # 为环中的节点创建圆形布局
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3)
            
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (
                    center_x + radius * np.cos(angle),
                    center_y + radius * np.sin(angle)
                )
        
        # 绘制每个环
        for idx, cycle in enumerate(cycles):
            color = colors[idx]
            
            # 绘制环中的节点
            cycle_nodes = cycle
            nx.draw_networkx_nodes(
                G, all_pos, nodelist=cycle_nodes,
                node_size=600, node_color=[color], 
                edgecolors='black', linewidths=2
            )
            
            # 绘制环中的有向边（透明度较低）
            cycle_edges = []
            for i in range(len(cycle)):
                u = cycle[i]
                v = cycle[(i + 1) % len(cycle)]
                if u in matching and matching[u] == v:
                    cycle_edges.append((u, v))
            
            # 绘制有向匹配边（透明）
            nx.draw_networkx_edges(
                G, all_pos, edgelist=cycle_edges,
                edge_color=[color], width=3.0, alpha=0.2,  # 改为透明
                arrows=True, arrowstyle='-|>', arrowsize=20,
                connectionstyle='arc3,rad=0.0', 
                node_size=600
            )
            
            # 如果只有一个节点，绘制有向自环（透明）
            if len(cycle) == 1 and cycle[0] in matching and matching[cycle[0]] == cycle[0]:
                node = cycle[0]
                pos_node = all_pos[node]
                # 绘制自环圆圈（透明）
                circle = plt.Circle(
                    (pos_node[0] + 0.4, pos_node[1]), 0.25,
                    fill=False, color=color, linewidth=2.5, alpha=0.2
                )
                plt.gca().add_patch(circle)
                # 添加箭头表示方向（透明）
                arrow = plt.Arrow(
                    pos_node[0] + 0.4 + 0.25, pos_node[1],
                    0.15, 0, width=0.2, color=color, alpha=0.2
                )
                plt.gca().add_patch(arrow)
    
        # 绘制非树边（使用实线，不透明）
        if hasattr(self, 'non_tree_edges') and self.non_tree_edges:
            nx.draw_networkx_edges(
                G, all_pos, edgelist=self.non_tree_edges,
                edge_color='red', width=2.5, alpha=0.8,
                arrows=True, arrowstyle='-|>', arrowsize=18,
                connectionstyle='arc3,rad=0.15',  # 弯曲以便区分
                node_size=600, style='solid'
            )
        
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels, 
                               font_size=11, font_weight='bold')
        
        # 添加图例
        from matplotlib.lines import Line2D
        legend_elements = [
            Line2D([0], [0], color='black', linewidth=3.0, alpha=0.2, label='matching edges'),
        ]
        if hasattr(self, 'non_tree_edges') and self.non_tree_edges:
            legend_elements.append(
                Line2D([0], [0], color='red', linewidth=2.5, label='non-tree edges')
            )
        plt.legend(handles=legend_elements, loc='upper right', fontsize=10)
        
        # 添加标题
        non_tree_info = f", {len(self.non_tree_edges)} non-tree edges" if hasattr(self, 'non_tree_edges') and self.non_tree_edges else ""
        title_text = f"{title}\n{subtitle}\nfind(s) {len(cycles)} cycles{non_tree_info}"
        plt.title(title_text, fontsize=14, pad=20)
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout()
        

    def visualize_matching_with_clusters(self, edges, title, subtitle="", non_tight_edges=None, clusters=None):
        """
        可视化匹配（有向图多环形式），
        并叠加 self.cluster_edges (聚类后的边)，每个簇使用不同颜色。
        """
        
        # 1. 初始化 Figure
        # 增大画布尺寸以便容纳图例
        fig = plt.figure(figsize=(10, 8)) 
        self.figures.append(fig)
        ax = plt.gca() # 获取当前的坐标轴
        
        # 2. 构建匹配关系和寻找环
        # --- [这部分代码与您原来的 visualize_matching5 完全相同] ---
        matching = {}
        for i, j in edges:
            matching[i] = j
        
        visited = set()
        cycles = []
        
        for start in range(self.n):
            if start in visited:
                continue
            
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            
            if len(cycle) > 1 and current == start:
                cycles.append(cycle)
            elif len(cycle) == 1 and start in matching and matching[start] == start:
                cycles.append(cycle)  # 自环
        
        if not cycles:
            for i, j in edges:
                cycles.append([i, j] if i != j else [i])
                
        # 3. 创建图和布局
        # --- [这部分代码与您原来的 visualize_matching5 完全相同] ---
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 计算每个环的位置
        num_cycles = len(cycles)
        cols = int(np.ceil(np.sqrt(num_cycles)))
        rows = int(np.ceil(num_cycles / cols))
        
        all_pos = {}
        
        for idx, cycle in enumerate(cycles):
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3)
            
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (
                    center_x + radius * np.cos(angle),
                    center_y + radius * np.sin(angle)
                )
                
        # 4. 绘制环的节点和透明匹配边
        # --- [这部分代码与您原来的 visualize_matching5 完全相同] ---
        
        # 为每个环分配颜色 (用于节点和透明边)
        cycle_colors = plt.cm.tab10(np.linspace(0, 1, max(10, len(cycles))))
        
        for idx, cycle in enumerate(cycles):
            color = cycle_colors[idx % 10]
            
            # 绘制环中的节点
            cycle_nodes = cycle
            nx.draw_networkx_nodes(
                G, all_pos, nodelist=cycle_nodes,
                node_size=600, node_color=[color],  
                edgecolors='black', linewidths=2, ax=ax
            )
            
            # 绘制环中的有向边（透明度较低）
            cycle_edges = []
            for i in range(len(cycle)):
                u = cycle[i]
                v = cycle[(i + 1) % len(cycle)]
                if u in matching and matching[u] == v:
                    cycle_edges.append((u, v))
            
            # 绘制有向匹配边（透明）
            nx.draw_networkx_edges(
                G, all_pos, edgelist=cycle_edges,
                edge_color=[color], width=3.0, alpha=0.2,  # 透明
                arrows=True, arrowstyle='-|>', arrowsize=20,
                connectionstyle='arc3,rad=0.0',  
                node_size=600, ax=ax
            )
            
            # 绘制有向自环（透明）
            if len(cycle) == 1 and cycle[0] in matching and matching[cycle[0]] == cycle[0]:
                node = cycle[0]
                pos_node = all_pos[node]
                circle = plt.Circle(
                    (pos_node[0] + 0.4, pos_node[1]), 0.25,
                    fill=False, color=color, linewidth=2.5, alpha=0.2
                )
                ax.add_patch(circle)
                arrow = plt.Arrow(
                    pos_node[0] + 0.4 + 0.25, pos_node[1],
                    0.15, 0, width=0.2, color=color, alpha=0.2
                )
                ax.add_patch(arrow)
    
        # 5. 【【【***关键修改点***】】】
        #    绘制聚类后的边 (从 self.clusters_result 读取)
        
        legend_elements = [
            Line2D([0], [0], color='black', linewidth=3.0, alpha=0.2, label='matching edges (Cycles)'),
        ]
        
        cluster_info = "" 
        
        # 检查 self.clusters_result 是否存在
        if hasattr(self, 'clusters_result') and self.clusters_result:
            num_clusters = len(self.clusters_result) # 或者直接使用 self.num_clusters
            #num_clusters = self.num_clusters
            
            cluster_colors = plt.cm.get_cmap('hsv', num_clusters + 1) 
            cluster_info = f", {num_clusters} clusters"
            
            # 遍历 self.clusters_result
            for idx, cluster in enumerate(self.clusters_result):
                if not cluster: 
                    continue
                    
                color = cluster_colors(idx / num_clusters)
                label = f'Cluster {idx} ({len(cluster)} edges)'
                
                nx.draw_networkx_edges(
                    G, all_pos,
                    edgelist=cluster,
                    edge_color=[color],
                    width=2.5, alpha=0.9,
                    arrows=True, arrowstyle='-|>', arrowsize=18,
                    connectionstyle='arc3,rad=0.15', 
                    node_size=600, style='solid', ax=ax
                )
                legend_elements.append(
                    Line2D([0], [0], color=color, linewidth=2.5, label=label)
                )
                
        else:
            # 回退：如果 self.clusters_result 还没有被计算
            cluster_info = ", (Clustering didn't run)"
            # 仍然尝试绘制 self.non_tree_edges (如果存在)
            if hasattr(self, 'non_tree_edges') and self.non_tree_edges:
                nx.draw_networkx_edges(
                    G, all_pos, edgelist=self.non_tree_edges,
                    edge_color='red', width=2.5, alpha=0.8,
                    arrows=True, arrowstyle='-|>', arrowsize=18,
                    connectionstyle='arc3,rad=0.15',
                    node_size=600, style='solid', ax=ax
                )
                legend_elements.append(
                    Line2D([0], [0], color='red', linewidth=2.5, label='non-tree edges (Unclustered)')
                )
    
        # 6. 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels,  
                                font_size=11, font_weight='bold', ax=ax)
        
        # 7. 添加图例和标题
        ax.legend(handles=legend_elements, loc='upper left', 
                  fontsize=9, bbox_to_anchor=(1.02, 1.0))
        
        title_text = f"{title}\n{subtitle}\nfind(s) {len(cycles)} cycles{cluster_info}"
        plt.title(title_text, fontsize=14, pad=20)
        
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout(rect=[0, 0, 0.85, 1])

    
    def visualize_cycle(self, edges, non_tree_set):
        """可视化有效环（有向图版本）"""
        fig = plt.figure(figsize=(2, 1))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', 
                              style='dashed', alpha=0.3)
    
        # 添加环中的边
        for i, j in edges:
            if (i, j) in non_tree_set:
                # 非树边（绿色虚线）
                G.add_edge(i, j, color='green', width=2.5, style='dashed')
            else:
                # 树边（蓝色实线）
                G.add_edge(i, j, color='blue', width=2.0)
    
        # 使用环形布局（适合显示环）
        pos = nx.circular_layout(G)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(
            G, pos, edgelist=all_edges,
            edge_color='lightgray', width=0.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3
        )
    
        # 绘制树边（蓝色实线）
        tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'blue']
        nx.draw_networkx_edges(
            G, pos, edgelist=tree_edges,
            edge_color='blue', width=2.0, 
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 绘制非树边（绿色虚线）
        non_tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'green']
        nx.draw_networkx_edges(
            G, pos, edgelist=non_tree_edges,
            edge_color='green', width=2.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 添加标题和图例
        non_tree_count = len([e for e in edges if e in self.non_tree_edges])
        plt.title(f"Valid cycles (non-tree edges: {non_tree_count})", fontsize=14)
    
        # 创建图例
        import matplotlib.patches as mpatches
        gray_patch = mpatches.Patch(color='lightgray', label='possbile edges')
        blue_patch = mpatches.Patch(color='blue', label='tree edges')
        green_patch = mpatches.Patch(color='green', label='non-tree edges')
        plt.legend(handles=[gray_patch, blue_patch, green_patch], loc='best')
    
        # 高亮显示环路径
        path_edges = []
        for i in range(len(edges)):
            path_edges.append(edges[i])
            if i < len(edges) - 1:
                # 添加中间节点
                pass
    
        # 添加环路径的动画效果（可选）
        for i, (u, v) in enumerate(path_edges):
            nx.draw_networkx_edges(
                G, pos, edgelist=[(u, v)],
                edge_color='red', width=3.0, 
                arrows=True, arrowstyle='-|>', arrowsize=20,
                alpha=0.7
            )
    
        plt.axis('off')
        plt.tight_layout()

    def visualize_valid_cycles(self, base_matching_edges, title, subtitle="", max_cycles_to_plot=100):
        """
        专门用于可视化 self.valid_cycles 的函数。
        
        【已更新】: 如果环的数量超过 'max_cycles_to_plot'，
        将从中随机抽取样本进行绘制，以防止崩溃。
        """
        
        # 0. 设置中文字体
        #set_chinese_font() 
        
        # 1. 检查是否存在 valid_cycles
        if not hasattr(self, 'valid_cycles') or not self.valid_cycles:
            msg = "visualize_valid_cycles: 未找到 'self.valid_cycles' 或其为空，无法绘图。"
            if hasattr(self, 'logger'): self.logger.warning(msg)
            else: print(f"警告: {msg}")
            return

        num_valid_cycles = len(self.valid_cycles)

        # 2. 【【核心修复】】: 随机抽样
        cycles_to_plot = []
        if num_valid_cycles > max_cycles_to_plot:
            msg = (f"警告: 有 {num_valid_cycles} 个环，数量过多。"
                   f"将随机抽取 {max_cycles_to_plot} 个进行显示。")
            if hasattr(self, 'logger'): self.logger.warning(msg)
            else: print(msg)
            
            # 从 17739 个环中随机抽取 100 个
            cycles_to_plot = random.sample(self.valid_cycles, max_cycles_to_plot)
        else:
            # 数量不多，全部绘制
            cycles_to_plot = self.valid_cycles

        num_to_plot = len(cycles_to_plot)
        
        # 3. 初始化 Figure
        fig = plt.figure(figsize=(12, 9)) 
        self.figures.append(fig)
        ax = plt.gca()
        
        # 4. 【关键】重用您的布局逻辑来计算 all_pos
        # ... (这部分代码与您之前的一样，用于计算 all_pos) ...
        matching = {}
        for i, j in base_matching_edges:
            matching[i] = j
        visited = set()
        base_cycles = []
        for start in range(self.n):
            if start in visited: continue
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            if len(cycle) > 1 and current == start:
                base_cycles.append(cycle)
            elif len(cycle) == 1 and start in matching and matching[start] == start:
                base_cycles.append(cycle)
        if not base_cycles:
            for i, j in base_matching_edges:
                base_cycles.append([i, j] if i != j else [i])
        
        num_base_cycles = len(base_cycles)
        cols = int(np.ceil(np.sqrt(num_base_cycles)))
        rows = int(np.ceil(num_base_cycles / cols))
        all_pos = {}
        for idx, cycle in enumerate(base_cycles):
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3)
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (center_x + radius * np.cos(angle), center_y + radius * np.sin(angle))
        
        
        # 5. 绘制底图 (灰色、半透明)
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        nx.draw_networkx_nodes(G, all_pos, nodelist=range(self.n), 
                               node_color='lightgray', node_size=500, alpha=0.6, ax=ax)
        nx.draw_networkx_edges(G, all_pos, edgelist=base_matching_edges,
                               edge_color='gray', width=2.0, alpha=0.3, arrows=False, ax=ax)
        
        # 6. 【核心】高亮绘制【抽样后】的 self.valid_cycles
        
        # 【警告修复】: 使用新的 Matplotlib API
        # cmap = plt.cm.get_cmap('hsv') # 旧的
        cmap = matplotlib.colormaps.get_cmap('hsv') # 新的
        
        legend_elements = [Line2D([0], [0], color='gray', alpha=0.3, lw=2, label='Background matching'),]

        # 遍历抽样后的环 (cycles_to_plot)，而不是所有的环
        for idx, cycle_edges in enumerate(cycles_to_plot):
            if not cycle_edges: continue
                
            color = cmap(idx / num_to_plot) # 使用 0 到 1 之间的值获取颜色
            nodes_in_cycle = list(set([u for u, v in cycle_edges] + [v for u, v in cycle_edges]))
            
            nx.draw_networkx_edges(
                G, all_pos, edgelist=cycle_edges,
                edge_color=[color], width=3.0, alpha=0.9,
                arrows=True, arrowstyle='-|>', arrowsize=18,
                connectionstyle='arc3,rad=0.1', node_size=600, ax=ax
            )
            nx.draw_networkx_nodes(
                G, all_pos, nodelist=nodes_in_cycle,
                node_color=[color], node_size=600,
                edgecolors='black', linewidths=1.0, ax=ax
            )
            
            if idx < 5:
                legend_elements.append(
                    Line2D([0], [0], color=color, lw=3, label=f'valid cycles (sample {idx})')
                )
            elif idx == 5:
                legend_elements.append(
                    Line2D([0], [0], color='black', marker='.', linestyle='None', 
                           label='... (and other sampling cycles)')
                )

        # 7. 绘制标签和设置
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels,  
                                font_size=11, font_weight='bold', ax=ax)
        
        ax.legend(handles=legend_elements, loc='upper left', 
                  fontsize=9, bbox_to_anchor=(1.02, 1.0))
        
        # 【中文修复】: 现在标题可以正常显示中文了
        title_text = (f"{title}\n{subtitle}\n"
                      f"highlights {num_to_plot} valid cycles (total number:{num_valid_cycles} )")
        plt.title(title_text, fontsize=14, pad=20)
        
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout(rect=[0, 0, 0.85, 1])

    def visualize_individual_cycles(self, base_matching_edges, num_to_plot=5, title_prefix="valid cycles"):
        """
        【新函数】为 self.valid_cycles 中的N个样本环，每个环创建一张独立的图。
        
        :param base_matching_edges: 用于计算布局和绘制底图的边 (例如 self.ap_edges)
        :param num_to_plot: 您想要绘制的独立图的数量 (例如 5)
        :param title_prefix: 每个图的标题前缀
        """
        
        # 0. 设置中文字体
        #set_chinese_font() 
        
        # 1. 检查和抽样
        if not hasattr(self, 'valid_cycles') or not self.valid_cycles:
            msg = "visualize_individual_cycles: 未找到 'self.valid_cycles' 或其为空。"
            if hasattr(self, 'logger'): self.logger.warning(msg)
            else: print(f"警告: {msg}")
            return

        num_total_cycles = len(self.valid_cycles)
        
        # 确保抽样数量不超过总数
        num_to_plot = min(num_to_plot, num_total_cycles)
        
        if num_to_plot == 0:
            return # 没什么可画的
            
        msg = f"总共有 {num_total_cycles} 个有效环。正在随机抽取 {num_to_plot} 个进行单独显示..."
        if hasattr(self, 'logger'): self.logger.info(msg)
        else: print(msg)
        
        # 随机抽样
        cycles_to_plot = random.sample(self.valid_cycles, num_to_plot)
        
        
        # 2. 【关键】计算一次所有节点的位置 (all_pos)
        # 我们必须在循环外计算它，以保证所有图的布局一致！
        # --- [这部分代码与您原来的 visualize_valid_cycles 布局逻辑完全相同] ---
        matching = {}
        for i, j in base_matching_edges:
            matching[i] = j
        visited = set()
        base_cycles = []
        for start in range(self.n):
            if start in visited: continue
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            if len(cycle) > 1 and current == start:
                base_cycles.append(cycle)
            elif len(cycle) == 1 and start in matching and matching[start] == start:
                base_cycles.append(cycle)
        if not base_cycles:
            for i, j in base_matching_edges:
                base_cycles.append([i, j] if i != j else [i])
        
        num_base_cycles = len(base_cycles)
        cols = int(np.ceil(np.sqrt(num_base_cycles)))
        rows = int(np.ceil(num_base_cycles / cols))
        all_pos = {}
        for idx, cycle in enumerate(base_cycles):
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3)
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (center_x + radius * np.cos(angle), center_y + radius * np.sin(angle))
        # --- [布局计算结束] ---


        # 3. 【核心】循环创建 N 个独立的图
        
        for i, cycle_edges in enumerate(cycles_to_plot):
            
            # (A) 为这个环创建一张新图
            fig = plt.figure(figsize=(8, 7)) # 8x7 的尺寸适合单个图
            self.figures.append(fig)
            ax = plt.gca()
            
            G = nx.DiGraph()
            G.add_nodes_from(range(self.n))

            # (B) 绘制灰色底图
            # 绘制所有节点 (灰色)
            nx.draw_networkx_nodes(
                G, all_pos, nodelist=range(self.n), 
                node_color='lightgray', node_size=500, alpha=0.6, ax=ax
            )
            # 绘制所有底图边 (灰色)
            nx.draw_networkx_edges(
                G, all_pos, edgelist=base_matching_edges,
                edge_color='gray', width=2.0, alpha=0.3, 
                arrows=False, ax=ax
            )

            # (C) 用红色高亮显示这一个环
            if not cycle_edges: continue # 跳过可能的空环

            nodes_in_cycle = list(set([u for u, v in cycle_edges] + [v for u, v in cycle_edges]))
            
            # 绘制高亮的边
            nx.draw_networkx_edges(
                G, all_pos,
                edgelist=cycle_edges,
                edge_color='red', # 统一用红色高亮
                width=3.0,
                alpha=1.0,
                arrows=True, arrowstyle='-|>', arrowsize=20,
                connectionstyle='arc3,rad=0.1',
                node_size=600,
                ax=ax
            )
            # 绘制高亮的节点
            nx.draw_networkx_nodes(
                G, all_pos,
                nodelist=nodes_in_cycle,
                node_color='red', # 统一用红色高亮
                node_size=600,
                edgecolors='black',
                linewidths=1.5,
                ax=ax
            )
            
            # (D) 绘制标签和标题
            labels = {i: f"{i}" for i in range(self.n)}
            nx.draw_networkx_labels(G, all_pos, labels=labels,  
                                    font_size=11, font_weight='bold', ax=ax)

            plt.title(f"{title_prefix} (Sample {i + 1} / {num_to_plot})\n(total number{num_total_cycles})", 
                      fontsize=14)
            plt.axis('off')
            plt.axis('equal')
            plt.tight_layout()

        
        msg = f"已为 {num_to_plot} 个环样本创建了独立的图像。"
        if hasattr(self, 'logger'): self.logger.info(msg)
        else: print(msg)

    # ... (您的类的其他函数) ...

    def visualize_tour2(self, edges, title, subtitle=""):
        """
        【新函数】
        专门用于可视化一个(假定的)哈密顿环 (Hamiltonian tour).
        它使用与其他可视化函数一致的布局逻辑。
        """
        
        # 0. 设置字体
        #set_chinese_font() 
        
        # 1. 设置 Figure
        fig = plt.figure(figsize=(8, 8)) # 方形图更适合环
        self.figures.append(fig)
        ax = plt.gca()
        
        # 2. 构建图 G
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 3. 【关键】计算布局
        # 我们重用 visualize_matching5 的布局逻辑，
        # 因为一个哈密顿环 *就是* 一个包含 n 个节点的、单独的环。
        # 该布局逻辑可以完美处理这种情况。
        
        matching = {}
        for i, j in edges:
            matching[i] = j
        
        visited = set()
        cycles = []
        # 查找环 (这里应该只找到一个环)
        for start in range(self.n):
            if start in visited: continue
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            if len(cycle) > 1 and current == start:
                cycles.append(cycle)
        
        # --- 布局计算 (与 visualize_matching5 相同) ---
        num_cycles = len(cycles)
        cols = int(np.ceil(np.sqrt(num_cycles)))
        rows = int(np.ceil(num_cycles / cols))
        all_pos = {}
        
        for idx, cycle in enumerate(cycles):
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            
            n_nodes = len(cycle)
            # n_nodes 此时应该等于 self.n, 所以半径会很大
            radius = max(1.0, n_nodes * 0.3) 
            
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (
                    center_x + radius * np.cos(angle),
                    center_y + radius * np.sin(angle)
                )
        # --- 布局计算结束 ---

        # 4. 绘制这个环
        
        # 绘制节点 (使用 "green" 绿色来表示这是一个好的"解")
        nx.draw_networkx_nodes(
            G, all_pos, nodelist=range(self.n),
            node_color='green', 
            node_size=600,
            edgecolors='black', linewidths=1.5, ax=ax,
            alpha=0.8
        )
        
        # 绘制边
        nx.draw_networkx_edges(
            G, all_pos, edgelist=edges,
            edge_color='green',
            width=3.0,
            alpha=0.8,
            arrows=True, arrowstyle='-|>', arrowsize=20,
            node_size=600,
            ax=ax
        )
        
        # 5. 绘制标签和标题
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels,
                                font_size=11, font_weight='bold', ax=ax)
        
        title_text = f"{title}\n{subtitle}"
        plt.title(title_text, fontsize=14, pad=20)
        
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout()

    def visualize_loops(self, edges, title, subtitle="", node_color='blue'):
        """
        【已修改/重命名】
        专门用于可视化一个“匹配”(边集), 它可能由一个或多个环组成。
        (例如: 绿色用于哈密顿环, 红色用于失败的候选解)
        """
        
        # 0. 设置字体
        #set_chinese_font() 
        
        # 1. 设置 Figure
        fig = plt.figure(figsize=(8, 8)) 
        self.figures.append(fig)
        ax = plt.gca()
        
        # 2. 构建图 G
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 3. 计算布局
        # 这段布局逻辑 *已经* 可以正确处理多个子环的情况
        
        matching = {}
        for i, j in edges:
            matching[i] = j
        
        visited = set()
        cycles = [] # 查找所有的子环
        for start in range(self.n):
            if start in visited: continue
            cycle = []
            current = start
            while current not in visited:
                visited.add(current)
                cycle.append(current)
                if current in matching:
                    current = matching[current]
                else:
                    break
            if len(cycle) > 1 and current == start:
                cycles.append(cycle)
        
        # --- 布局计算 (与之前相同) ---
        num_cycles = len(cycles)
        # (如果找不到环, 至少要有一个网格)
        cols = int(np.ceil(np.sqrt(num_cycles))) if num_cycles > 0 else 1
        rows = int(np.ceil(num_cycles / cols)) if num_cycles > 0 else 1
        all_pos = {}
        
        for idx, cycle in enumerate(cycles):
            row = idx // cols
            col = idx % cols
            center_x = col * 4
            center_y = -row * 4
            
            n_nodes = len(cycle)
            radius = max(1.0, n_nodes * 0.3) 
            
            for i, node in enumerate(cycle):
                angle = 2 * np.pi * i / n_nodes - np.pi / 2
                all_pos[node] = (
                    center_x + radius * np.cos(angle),
                    center_y + radius * np.sin(angle)
                )
        
        # (如果有些节点不在任何环中, 给它们一个默认位置)
        for node in G.nodes:
            if node not in all_pos:
                all_pos[node] = (random.uniform(-1,1), random.uniform(-1,1))
        # --- 布局计算结束 ---

        # 4. 绘制这个环
        
        # 绘制节点 (使用传入的 node_color)
        nx.draw_networkx_nodes(
            G, all_pos, nodelist=range(self.n),
            node_color=node_color,  # <-- 使用参数
            node_size=600,
            edgecolors='black', linewidths=1.5, ax=ax,
            alpha=0.8
        )
        
        # 绘制边 (使用传入的 node_color)
        nx.draw_networkx_edges(
            G, all_pos, edgelist=edges,
            edge_color=node_color, # <-- 使用参数
            width=3.0,
            alpha=0.8,
            arrows=True, arrowstyle='-|>', arrowsize=20,
            node_size=600,
            ax=ax
        )
        
        # 5. 绘制标签和标题
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, all_pos, labels=labels,
                                font_size=11, font_weight='bold', ax=ax)
        
        # 【重要】在标题中加入子环的数量
        title_text = f"{title}\n{subtitle}\n(Graph inclues {num_cycles} subcycles)"
        plt.title(title_text, fontsize=14, pad=20)
        
        plt.axis('off')
        plt.axis('equal')
        plt.tight_layout()
        
        
    def visualize_tour(self, tour_edges, cost):
        """可视化ATSP环"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 添加边
        for i, j in tour_edges:
            G.add_edge(i, j, weight=self.C[i][j])
        
        # 创建环形布局
        pos = nx.circular_layout(G)
        
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=700, node_color='lightgreen')
        nx.draw_networkx_labels(G, pos, font_size=12, font_weight='bold')
        
        # 绘制边
        edge_labels = {(i, j): f"{self.C[i][j]:.2f}" for i, j in G.edges()}
        nx.draw_networkx_edges(G, pos, edge_color='blue', width=2.0, 
                              arrows=True, arrowsize=20, 
                              connectionstyle='arc3,rad=0.1')
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10)
        
        # 添加标题
        plt.title(f"ATSP optimal solutions\ntotal costs: {cost:.4f}", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def show_visualizations(self):
        """显示所有可视化结果"""
        print(f"显示 {len(self.figures)} 个可视化图表...")
        plt.show()

    
    def edges_to_tour(self):
        """
        将AP解边集转换为ATSP环路序列并计算路径成本
        :return: 成功返回路径成本，失败返回0
        """
        if not self.ap_edges:
            self.logger.error("边集不能为空")
            return 0
        
        # 构建邻接字典：{起点: 终点}
        adj_map = {}
        for u, v in self.ap_edges:
            if u in adj_map:
                self.logger.error(f"顶点 {u} 存在多条出边，非合法环路")
                return 0
            adj_map[u] = v
        
        # 验证顶点一致性
        all_vertices = set(adj_map.keys()) | set(adj_map.values())
        if len(all_vertices) != len(adj_map):
            self.logger.error("存在孤立顶点或度数异常")
            return 0
    
        # 从任意起点开始追踪路径
        start = next(iter(adj_map.keys()))
        tour = [start]
        current = start
        visited = set([start])
        
        # 沿邻接映射追踪路径
        while True:
            next_node = adj_map.get(current)
            if next_node is None:
                self.logger.error(f"顶点 {current} 无出边，路径中断")
                return 0
            if next_node in visited and next_node != start:
                self.logger.error(f"检测到子环：顶点 {next_node} 被重复访问")
                return 0
        
            # 回到起点时结束
            if next_node == start:
                tour.append(start)
                break
            
            tour.append(next_node)
            visited.add(next_node)
            current = next_node
    
        # 验证路径覆盖所有顶点
        if len(visited) != len(all_vertices):
            missing = all_vertices - set(tour)
            self.logger.error(f"未覆盖所有顶点，缺失顶点: {missing}")
            return 0
    
        # 计算路径总成本
        total_cost = 0
        for i in range(len(tour) - 1):
            from_node = tour[i]
            to_node = tour[i+1]
            total_cost += self.C[from_node][to_node]
        
        self.logger.info(f"转换成功: 路径 = {tour}, 成本 = {total_cost}")
        return total_cost  
    
    def solve(self):
        """执行完整算法"""
        print(f"solve {self.n} cities ATSP problem (k_max = {self.k_max})")
        
        # 步骤1: 求解AP
        self.solve_ap()
        #print(f"AP optimal cost: {self.ap_cost:.4f}")
        
        cycle = self.edges_to_tour()
        if cycle != 0:
            results_true.append(1)
            cost = cycle
            return None, cost
        if cycle == 0 and self.C.shape[0] > 151:
            results_true.append(0)
            percentage.append(0)
            return None, float('inf')
            
        if cycle == 0 and self.C.shape[0] <= 151:
            # step 2.1 get dual variables
            
            # 步骤2: 构建生成树T*
            self.build_spanning_tree()
            print(f"spaning tree T* construction completed ({len(self.T_edges)} edges)")
        
            # 步骤3: 识别非树边
            self.get_non_tree_edges()
            print(f"non-basic edges number: {len(self.non_tree_edges)}")
        
            # 步骤4: 枚举有效环
            self.find_valid_structures()
            #print(f"有效环: {self.valid_cycles}")
        
            # 步骤5: 生成候选解
            self.candidates = {}  # 重置候选解字典
            edges, cost = self.generate_candidates()
            if edges is not None:
                key = tuple(sorted(edges))
                self.candidates[key] = cost
            print(f"生成候选解数量: {len(self.candidates)}")
            print(f"候选解: {self.candidates}")

            # 步骤6: 找到最优ATSP解
            tour_edges, cost = self.find_optimal_tour()
        
            if tour_edges:
                print(f"🎯 found ATSP optimal solution: cost = {cost:.4f}")
                self.visualize_tour(tour_edges, cost)
                results_true.append(1)
                return tour_edges, cost
            else:
                print("未找到Hamilton环解，使用karp算法")
                results_true.append(1)
                #percentage.append(0)
                tour_edges, cost = karp_patching_atsp(self.C, verbose=False)
                #return None, float('inf')
                return None, cost
    
def process_cluster(cluster, bipartite_edges, n):
    """修正后的聚类处理函数"""
    valid_cycles_local = []
    
    # 增大k的搜索范围
    max_k = min(len(cluster), 10)  # 增加到10
    
    for k in range(1, max_k + 1):
        for non_tree_set in combinations(cluster, k):
            G_directed = nx.DiGraph()
            
            # 正确处理二分图边
            for (u, v) in bipartite_edges:
                # v应该是 j+n 形式，转换为正确的方向
                if v >= n:
                    actual_j = v - n
                    G_directed.add_edge(actual_j, u)  # B->A
                else:
                    G_directed.add_edge(v, u)
            
            # 添加非树边 A->B
            for (i, j) in non_tree_set:
                G_directed.add_edge(i, j)
            
            # 查找环
            try:
                for cycle_nodes in nx.simple_cycles(G_directed):
                    if len(cycle_nodes) % 2 != 0:
                        continue
                    
                    # 验证环的交替性
                    cycle_edges = []
                    is_valid = True
                    
                    for idx in range(len(cycle_nodes)):
                        u = cycle_nodes[idx]
                        v = cycle_nodes[(idx + 1) % len(cycle_nodes)]
                        
                        # 确保边在正确的方向
                        if idx % 2 == 0:  # 应该是A->B
                            if (u, v) in non_tree_set or any(
                                (u, w-n) == (u, v) for (_, w) in bipartite_edges if w >= n
                            ):
                                cycle_edges.append((u, v))
                            else:
                                is_valid = False
                                break
                        else:  # 应该是B->A
                            cycle_edges.append((v, u))  # 翻转存储
                    
                    if is_valid and len(cycle_edges) == len(cycle_nodes):
                        valid_cycles_local.append(cycle_edges)
                        
            except nx.NetworkXNoCycle:
                pass
    
    return list(set(map(tuple, valid_cycles_local)))


def generate_random_instance(n, dist_type='exp'):
    """Generate random ATSP inistances"""
    if dist_type == 'exp':
        # exponential dist
        C = np.random.exponential(scale=1.0, size=(n, n))
    elif dist_type == 'uniform':
        # Uniform dist
        C = np.random.uniform(0, 1, size=(n, n))
    else:
        # random dist P(C≤x)=x+O(x²)
        C = np.zeros((n, n))
        for i in range(n):
            for j in range(n):
                u = np.random.uniform(0, 1)
                C[i][j] = u - 0.5 * u**2
    
    # execpt for the diagonal
    np.fill_diagonal(C, np.random.uniform(0, 1, n))
    return C


def verify_solution(C, tour):
    """Verify the correctness of the solution"""
    cost = 0
    visited = set()
    current = 0
    
    for _ in range(len(C)):
        next_node = tour[current]
        cost += C[current][next_node]
        visited.add(next_node)
        current = next_node
    
    return len(visited) == len(C), cost


def parse_atsp_file(filename):
    """
    解析TSPLIB格式的ATSP文件，返回cost matrix
    
    参数:
        filename: ATSP文件路径
    
    返回:
        cost_matrix: numpy数组，形状为(n, n)的成本矩阵
        metadata: 字典，包含文件的元数据信息
    """
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    metadata = {}
    matrix_data = []
    in_matrix_section = False
    
    for line in lines:
        line = line.strip()
        
        # 跳过空行
        if not line:
            continue
        
        # 结束标记
        if line == 'EOF':
            break
        
        # 解析头部信息
        if ':' in line and not in_matrix_section:
            key, value = line.split(':', 1)
            metadata[key.strip()] = value.strip()
        
        # 检测矩阵数据段开始
        elif line == 'EDGE_WEIGHT_SECTION':
            in_matrix_section = True
        
        # 读取矩阵数据
        elif in_matrix_section:
            # 提取行中的所有数字
            numbers = re.findall(r'-?\d+', line)
            matrix_data.extend([int(num) for num in numbers])
    
    # 获取维度
    dimension = int(metadata.get('DIMENSION', 0))
    
    if dimension == 0:
        raise ValueError("未找到DIMENSION信息")
    
    # 验证数据数量
    expected_count = dimension * dimension
    if len(matrix_data) != expected_count:
        raise ValueError(f"数据数量不匹配：期望{expected_count}个，实际{len(matrix_data)}个")
    
    # 转换为numpy矩阵
    cost_matrix = np.array(matrix_data).reshape(dimension, dimension)
    
    # 将9999替换为一个大数值（表示不可达）
    #cost_matrix = cost_matrix.astype(float)
    #cost_matrix[cost_matrix == 9999] = np.inf
    
    return cost_matrix, metadata



# ===================== Main program =====================
def main(n):
    #np.random.seed(42)
    
    #print("="*50)
    #n = 17
    print("Bell & Frieze (2024) ATSP algorithm implementation")
    #print("="*50)

    try:
        C, metadata = parse_atsp_file('br171.atsp')
        
        print("=== 文件元数据 ===")
        for key, value in metadata.items():
            print(f"{key}: {value}")
        
        print(f"\n=== Cost Matrix ===")
        print(f"形状: {C.shape}")
        print(f"\n矩阵内容:")
        print(C)
        
        print(f"\n=== 统计信息 ===")
        # 排除无穷大值的统计
        finite_values = C[np.isfinite(C)]
        print(f"最小值: {finite_values.min()}")
        print(f"最大值: {finite_values.max()}")
        print(f"平均值: {finite_values.mean():.2f}")
        
    except FileNotFoundError:
        print("文件未找到，请确保br17.atsp在当前目录下")
    except Exception as e:
        print(f"解析错误: {e}")
        
    # Generate a random instance (n=5)
    #n = n
    print(f"\nGenrate a {n} cities random ATSP instance...")
    C = generate_random_instance(n, dist_type='uniform')
    for i in range(n):
        C[i][i] = 1000000
    #print("Cost matrix:")
    #print(np.round(C, 3))
    #C = C.tolist()
    #print(C2)
    
    # Generate a solver
    solver = BellFriezeATSP(C)
    
    # solve ATSP
    print("\nBegin to solve...")
    start = time.perf_counter()
    
    tour_edges, cost = solver.solve()

    end = time.perf_counter()
    timecount.append(end - start)
    del start, end
    
    if tour_edges:
        # transform the list of edges into a tour
        graph = {i: j for (i, j) in tour_edges}
        tour_path = [0]
        current = 0
        for _ in range(n-1):
            current = graph[current]
            tour_path.append(current)
        
        # Verify and print the tour
        valid, verified_cost = verify_solution(C, graph)
        print(f"Otpimal tour: {tour_path}")
        print(f"Verification: {'Sucess' if valid else 'failed'}, cost={verified_cost:.4f}")
        actual_cost = solve_tsp_from_matrix(C, show_plot=False)
        percentage.append(verified_cost/actual_cost)
        del actual_cost, verified_cost
    else:
        actual_cost = solve_tsp_from_matrix(C, show_plot=False)
        percentage.append(cost/actual_cost)
        del actual_cost, cost
        
        # Show all of the visualization
        #solver.show_visualizations()
```

```python
import numpy as np
from scipy import stats

def calculate_confidence_interval(data, confidence=0.95):
    """
    为一个给定的数值数据集计算置信区间。

    参数:
    data (list or np.array): 包含实验数据的列表或Numpy数组。
    confidence (float): 期望的置信水平，默认为0.95 (95%)。

    返回:
    tuple: 包含样本均值、置信区间下限和置信区间上限的元组。
    """
    # 将输入数据转换为Numpy数组以便计算
    data_array = np.array(data)
    n = len(data_array)

    # 如果数据点少于2个，则无法计算置信区间
    if n < 2:
        print("数据样本太少，无法计算置信区间。")
        # 返回均值，置信区间为无效值
        return np.mean(data_array), float('nan'), float('nan')

    # 1. 计算样本均值
    mean = np.mean(data_array)

    # 2. 计算样本标准差 (使用 n-1 作为分母，ddof=1)
    std_dev = np.std(data_array, ddof=1)

    # 3. 获取临界值
    # 我们使用t分布，因为它对小样本和大样本都适用
    # ppf 是百分点函数 (Percent Point Function)，用于计算给定概率下的t分数
    # 对于95%的置信度，我们需要覆盖中间95%的区域，所以在分布的两尾各留出 2.5%
    # 因此我们查找的点是 1 - (alpha / 2) = 1 - (0.05 / 2) = 0.975
    alpha = 1 - confidence
    t_critical = stats.t.ppf(1 - alpha / 2, df=n - 1)

    # 4. 计算误差范围 (Margin of Error)
    # 误差范围 = 临界值 * (标准差 / 样本量的平方根)
    margin_of_error = t_critical * (std_dev / np.sqrt(n))

    # 5. 计算置信区间的上下限
    lower_bound = mean - margin_of_error
    upper_bound = mean + margin_of_error

    return mean, lower_bound, upper_bound

# --- 使用示例 ---

# 1. 假设这是您收集到的30次实验的【运算时间】数据 (单位：秒)
#    请将这里的模拟数据换成您的真实数据！
computation_times = timecount
#computation_times = [5.2, 4.8, 5.5, 5.1, 4.9, 5.3, 5.8, 4.7, 5.4, 5.2,
                     #4.9, 5.0, 5.6, 5.3, 5.1, 5.4, 5.2, 4.8, 5.7, 5.0,
                     #5.1, 5.5, 5.3, 4.9, 5.2, 5.4, 5.0, 5.3, 5.1, 5.6]

# 2. 假设这是您收集到的30次实验的【近似比 q】数据
#    请将这里的模拟数据换成您的真实数据！
approximation_ratios = percentage
#approximation_ratios = [1.12, 1.15, 1.09, 1.20, 1.11, 1.14, 1.16, 1.08, 1.13, 1.15,
                        #1.17, 1.10, 1.12, 1.14, 1.18, 1.09, 1.11, 1.16, 1.13, 1.15,
                        #1.19, 1.12, 1.14, 1.10, 1.13, 1.15, 1.17, 1.08, 1.11, 1.16]


# --- 计算并打印结果 ---

print("--- Runing time analysis ---")
time_mean, time_lower, time_upper = calculate_confidence_interval(computation_times)
if not np.isnan(time_lower):
    print(f"Sample size (n): {len(computation_times)}")
    print(f"Average runing time: {time_mean:.4f} second(s)")
    print(f"95% Confidence interval: [{time_lower:.4f}, {time_upper:.4f}]")
    print(f"Description: We have 95% of confidence to belive that，the real average runing time of this algorithm is between {time_lower:.4f} and {time_upper:.4f} second(s)。")

print("\n" + "="*30 + "\n")

print("--- Approximation ratio analysis ---")
ratio_mean, ratio_lower, ratio_upper = calculate_confidence_interval(approximation_ratios)
if not np.isnan(ratio_lower):
    print(f"Sample size (n): {len(approximation_ratios)}")
    print(f"Average approximation ratio: {ratio_mean:.4f}")
    print(f"95% Confidence interval: [{ratio_lower:.4f}, {ratio_upper:.4f}]")
    print(f"Description: We have 95% of confidence to belive that，the real average approximation ratio of this algorithm is between {ratio_lower:.4f} and {ratio_upper:.4f} .")
```

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

def clopper_pearson_interval(k, n, alpha=0.05):
    """
    Clopper-Pearson精确置信区间（适用于小样本）
    :param k: 成功次数
    :param n: 总实验次数
    :param alpha: 显著性水平（默认95%置信水平）
    :return: (lower_bound, upper_bound)
    """
    lower = stats.beta.ppf(alpha/2, k, n - k + 1)
    upper = stats.beta.ppf(1 - alpha/2, k + 1, n - k)
    return max(0, lower), min(1, upper)  # 限制在[0,1]范围内

def wilson_interval(k, n, alpha=0.05):
    """
    Wilson置信区间（推荐通用场景）
    :param k: 成功次数
    :param n: 总实验次数
    :param alpha: 显著性水平
    :return: (lower_bound, upper_bound)
    """
    p = k / n
    z = stats.norm.ppf(1 - alpha/2)
    denominator = 1 + z**2 / n
    centre = (p + z**2/(2*n)) / denominator
    margin = z * np.sqrt(p*(1-p)/n + z**2/(4*n**2)) / denominator
    return centre - margin, centre + margin

def bootstrap_success_interval(success_flags, n_bootstrap=10000, alpha=0.05):
    """
    Bootstrap置信区间（适用于复杂指标）
    :param success_flags: 布尔数组，True表示成功
    :param n_bootstrap: 重采样次数
    :param alpha: 显著性水平
    :return: (lower_bound, upper_bound)
    """
    success_rates = []
    for _ in range(n_bootstrap):
        # 有放回重采样
        resampled = np.random.choice(success_flags, size=len(success_flags), replace=True)
        success_rates.append(np.mean(resampled))
    return np.percentile(success_rates, 100*alpha/2), np.percentile(success_rates, 100*(1-alpha/2))

# ========== 示例使用 ==========
if __name__ == "__main__":
    # 模拟实验数据（100次运行，成功80次）
    # np.random.seed(42)  # 固定随机种子
    # n_trials = 100
    n_trials = 50
    success_count = count
    
    success_flags = np.array([True]*success_count + [False]*(n_trials - success_count))

    # 方法1：Clopper-Pearson（精确方法）
    cp_low, cp_high = clopper_pearson_interval(success_count, n_trials)
    
    # 方法2：Wilson区间
    wilson_low, wilson_high = wilson_interval(success_count, n_trials)
    
    # 方法3：Bootstrap（10000次重采样）
    bs_low, bs_high = bootstrap_success_interval(success_flags)

    # 输出结果对比
    #print(f"Clopper-Pearson (保守): [{cp_low:.4f}, {cp_high:.4f}]")
    print(f"Wilson confidence Interval:        [{wilson_low:.4f}, {wilson_high:.4f}]")
    #print(f"Bootstrap (10k iter):   [{bs_low:.4f}, {bs_high:.4f}]")
    print(f"Optimal rate: {success_count/n_trials:.4f}")

    # 理论界限对比（假设论文中理论下限为0.70）
    theoretical_bound = 0.94227001848
    plt.figure(figsize=(10, 6))
    plt.errorbar(x=[0], y=[success_count/n_trials], 
                 yerr=[[success_count/n_trials - wilson_low], [wilson_high - success_count/n_trials]], 
                 fmt='o', label='Wilson interval', capsize=6)
    plt.axhline(y=theoretical_bound, color='r', linestyle='--', label='Theoretical Bound (Paper)')
    plt.title('95% Probability Bound vs Theoretical Bound')
    plt.ylabel('Success Probability')
    plt.legend()
    plt.grid(axis='y')
    plt.show()
```
