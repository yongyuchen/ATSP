```python
# 基于成本矩阵的 TSP 求解程序
from pyscipopt import Model, Expr
import numpy as np
import matplotlib.pyplot as plt
from os import path

def add_mtz_subtour_constraints(n, model, x):
    """MTZ 子路径消除约束（精简版）"""
    # 创建辅助变量
    u = [model.addVar(lb=0, vtype="C", name=f"u_{i}") for i in range(n)]
    
    # 添加 MTZ 约束
    for i in range(1, n):
        for j in range(1, n):
            model.addCons(u[i] - u[j] + (n - 1) * x[i][j] <= n - 2)
        model.addCons(u[i] >= 1)
        model.addCons(u[i] <= n - 1)
    
    # 固定起点约束
    model.addCons(u[0] == 0)

def solve_tsp_from_matrix(cost_matrix, node_coords=None, show_plot=True):
    """
    基于成本矩阵的 TSP 求解函数
    
    cost_matrix: 成本矩阵 (n x n)
    node_coords: 节点坐标列表 (可选)
    show_plot: 是否显示图形结果 (需要坐标)
    """
    # 1. 获取问题维度
    n = len(cost_matrix)
    print(f"TSP 问题: {n}节点")
    
    # 2. 创建模型
    model = Model("tsp_matrix")
    
    # 3. 创建变量
    x = [[model.addVar(lb=0, ub=1, vtype="B", name=f"x_{i}_{j}") 
          for j in range(n)]
          for i in range(n)]
    
    # 4. 设置目标函数
    obj = Expr()
    for i in range(n):
        for j in range(n):
            if i != j:
                obj += cost_matrix[i][j] * x[i][j]
    model.setObjective(obj, "minimize")
    
    # 5. 添加基本约束
    for i in range(n):
        incoming = sum(x[j][i] for j in range(n))
        outgoing = sum(x[i][j] for j in range(n))
        model.addCons(incoming == 1, name=f"in_{i}")
        model.addCons(outgoing == 1, name=f"out_{i}")
        model.addCons(x[i][i] == 0, name=f"self_{i}")
    
    # 6. 添加 MTZ 约束
    add_mtz_subtour_constraints(n, model, x)
    
    # 7. 求解模型
    print("求解中...")
    model.hideOutput()  # 隐藏求解器详细输出
    model.optimize()
    
    # 8. 显示结果
    if model.getNSols() > 0:
        sol = model.getBestSol()
        objval = model.getObjVal(sol)
        
        # 提取解
        xval = [[model.getSolVal(sol, x[i][j]) 
                 for j in range(n)]
                 for i in range(n)]
        
        # 获取路径
        tour = []
        current = 0
        while len(tour) < n:
            tour.append(current)
            for j in range(n):
                if xval[current][j] > 0.5:
                    current = j
                    break
        tour.append(tour[0])
        
        # 计算实际成本
        actual_cost = 0
        for i in range(len(tour)-1):
            actual_cost += cost_matrix[tour[i]][tour[i+1]]
        
        # 显示结果
        print(f"\n最优成本: {actual_cost:.2f}")
        print(f"求解时间: {model.getTotalTime():.1f}秒")
        print(f"路径: {' → '.join(map(str, tour))}")
        return actual_cost
        
        # 显示图形 (如果有坐标)
        if show_plot and node_coords is not None and len(node_coords) == n:
            try:
                # 提取坐标
                coords = node_coords
                
                # 绘制图形
                plt.figure(figsize=(10, 8))
                
                # 绘制节点
                x_vals = [coords[node][0] for node in tour]
                y_vals = [coords[node][1] for node in tour]
                plt.plot(x_vals, y_vals, 'b-', marker='o', markersize=4)
                
                # 标记起点和终点
                plt.plot(coords[tour[0]][0], coords[tour[0]][1], 'ro', markersize=8)
                plt.plot(coords[tour[-2]][0], coords[tour[-2]][1], 'go', markersize=8)
                
                # 添加标签
                for i, coord in enumerate(coords):
                    plt.text(coord[0], coord[1], str(i), fontsize=8, ha='center', va='center')
                
                plt.title(f'TSP 路径 - 成本: {actual_cost:.2f}')
                plt.xlabel('X坐标')
                plt.ylabel('Y坐标')
                plt.grid(True, linestyle='--', alpha=0.3)
                plt.show()
            except Exception as e:
                print(f"图形显示错误: {str(e)}")
        elif show_plot and node_coords is None:
            print("未提供节点坐标，无法显示图形")
    else:
        print("未找到可行解")
```

```python
import numpy as np
from scipy.optimize import linear_sum_assignment
%matplotlib inline
import matplotlib.pyplot as plt
import networkx as nx
import math
from itertools import combinations
import matplotlib.colors as mcolors
from scipy.sparse.csgraph import connected_components
from scipy.sparse import csr_matrix
from pyscipopt import Model  # 添加SCIP求解器
from concurrent.futures import ThreadPoolExecutor
import logging
# 使用更高效的并行处理框架
from joblib import Parallel, delayed
# 使用近似聚类替代精确社区检测
from sklearn.cluster import DBSCAN

# 配置日志
logging.basicConfig(level=logging.INFO, 
                    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')

class BellFriezeATSP:
    __slots__ = ['C', 'n', 'ap_edges', 'ap_cost', 'uu', 'vv', 'non_tree_edges', 'candidates', 'debug', 'logger', 'T_star', 'valid_cycles', 
                'figures', 'k_max', 'base_graph', 'base_graph_edges', 'T_edges', 'bipartite_edges']  # 仅声明必需字段
    def __init__(self, cost_matrix, debug=False):
        # 输入验证
        if not isinstance(cost_matrix, np.ndarray):
            raise TypeError("成本矩阵必须为NumPy数组")
        if cost_matrix.shape[0] != cost_matrix.shape[1]:
            raise ValueError("成本矩阵必须为方阵")
        if np.any(cost_matrix < 0):
            raise ValueError("成本不能为负值")
            
        #super().__init__(*args, **kwargs)
        self.C = cost_matrix
        self.n = len(cost_matrix)
        #self.spatial_threshold = 0.3 * self.n
        self.debug = debug
        self.logger = logging.getLogger('ATSP-Solver')
        self.logger.info(f"初始化ATSP求解器，城市数: {self.n}")
        
        # 算法数据结构
        self.ap_edges = None
        self.ap_cost = None
        # 使用边列表代替完整的networkx图对象
        self.T_edges = set()  # 存储生成树边(i, j)而不是完整的图对象
        #self.T_star = None
        self.non_tree_edges = None
        self.valid_cycles = None
        self.bipartite_edges = None
        self.candidates = {}
        #self.compute_new_edges = {}
        self.figures = []  # store images
        
        # 对偶变量
        self.uu = np.zeros(self.n)
        self.vv = np.zeros(self.n)
        
        # 优化参数：限制非树边组合大小
        self.k_max = min(5, max(1, math.ceil(math.log(self.n)**2.5)))
        self.logger.debug(f"设置k_max={self.k_max}")
        
    def solve_ap(self):
        """使用SCIP求解指派问题并获取对偶变量"""
        n = self.n
        model = Model("AssignmentProblem")
        
        # 创建变量
        x = {}
        for i in range(n):
            for j in range(n):
                x[i, j] = model.addVar(vtype="C", lb=0, ub=1, name=f"x_{i}_{j}")
        
        # 设置目标函数：最小化总成本
        model.setObjective(sum(self.C[i, j] * x[i, j] for i in range(n) for j in range(n)), "minimize")
        
        # 添加行约束：每行和为1
        row_constraints = []
        for i in range(n):
            cons = model.addCons(sum(x[i, j] for j in range(n)) == 1, name=f"row_{i}")
            row_constraints.append(cons)
        
        # 添加列约束：每列和为1
        col_constraints = []
        for j in range(n):
            cons = model.addCons(sum(x[i, j] for i in range(n)) == 1, name=f"col_{j}")
            col_constraints.append(cons)
        
        # 求解模型
        model.optimize()
        
        # 检查求解状态
        if model.getStatus() != "optimal":
            raise RuntimeError("SCIP failed to find optimal solution")
        
        # 保存最优匹配边
        self.ap_edges = []
        for i in range(n):
            for j in range(n):
                if model.getVal(x[i, j]) > 0.5:  # 大于0.5视为选中
                    self.ap_edges.append((i, j))

        # 计算并存储AP成本
        self.ap_cost = model.getObjVal()
        print(f"AP最优成本: {self.ap_cost:.4f}")
        
        print("ap edges")
        print(self.ap_edges)
        # 提取对偶变量
        self.uu = np.zeros(n)
        self.vv = np.zeros(n)
        
        for i in range(n):
            self.uu[i] = model.getDualsolLinear(row_constraints[i])
        
        for j in range(n):
            self.vv[j] = model.getDualsolLinear(col_constraints[j])
            
        # 添加对偶验证
        dual_sum = sum(self.uu) + sum(self.vv)
        if abs(self.ap_cost - dual_sum) > 1e-5:
            self.logger.warning(f"对偶间隙: {abs(self.ap_cost - dual_sum):.6f}")
    
    def build_spanning_tree(self):
        """并行构建生成树"""
        tol = 1e-5
        self.logger.info("开始构建生成树...")

        
        # 并行识别紧边
        with ThreadPoolExecutor() as executor:
            results = list(executor.map(self._find_row_tight_edges, range(self.n)))
        tight_edges = [item for sublist in results for item in sublist]
        
        
        # 构建基础图（后续复用）
        self.base_graph = nx.Graph()
        parent = list(range(2 * self.n))
        T_edges = set(self.ap_edges)
        
        # ... 并查集操作不变 ...
        
        n = self.n
        
        # 步骤3: 合并匹配边
        for i, j in self.ap_edges:
            self.union(parent, i, j + n)
        
        # 步骤4: 添加非匹配紧边连接分量
        for (i, j) in tight_edges:
            if (i, j) in self.ap_edges:  # 跳过匹配边
                continue
                
            a_node = i
            b_node = j + n
            if self.find(parent, a_node) != self.find(parent, b_node):
                self.union(parent, a_node, b_node)
                T_edges.add((i, j))
                if len(T_edges) == 2 * n - 1:  # 已达2n-1条边
                    break

        # 添加到基础图
        #for (i, j) in T_edges:
            #self.base_graph.add_edge(i, j+self.n)
        #self.T_star = self.base_graph.copy()
        for (i, j) in T_edges:
        # 存储为二分图边 (i, j+n)
            self.T_edges.add((i, j + self.n))  # 关键修改
        self.bipartite_edges = list(self.T_edges)  # 用于后续环检测
        #for (i, j) in T_edges:
            #self.T_edges.add((i, j))  # 仅存储边元组
        #self.T_star = self.T_edges.copy()
        self.logger.info(f"生成树构建完成，边数: {len(T_edges)}")

        # 释放基础图内存
        #del self.base_graph
        #self.base_graph_edges = list(self.T_star.edges())  # 只存储边元组
        
        # 构建生成树图，使用整数节点
        #self.T_star = nx.Graph()
        #for (i, j) in T_edges:
            # 添加边：A部节点i，B部节点j+n
            #self.T_star.add_edge(i, j+self.n, weight=self.C[i, j])
    
        #return list(T_edges), M_star, u, v
        #self.T_star = T_edges
        #A = nx.to_numpy_array(self.T_star)
        #print("YYC spaning tree hrefdsjflsjakl'fjasfjaslfkafafa\n")
        #print(T_edges)
        #print(A)
        
        
        # Visualization the spaning tree
        #self.visualize_spanning_tree(T_star)
        #return T_edges

    # 并查集工具函数"
    @staticmethod
    def find(parent, x):
        """查找根节点（带路径压缩）"""
        if parent[x] != x:
            parent[x] = BellFriezeATSP.find(parent, parent[x])
        return parent[x]
    
    @staticmethod
    def union(parent, x, y):
        """合并两个集合"""
        root_x = BellFriezeATSP.find(parent, x)
        root_y = BellFriezeATSP.find(parent, y)
        if root_x != root_y:
            parent[root_y] = root_x
            
    def _find_row_tight_edges(self, i):
        """辅助函数：并行处理单行"""
        row = []
        for j in range(self.n):
            diff = abs(self.C[i, j] - (self.uu[i] + self.vv[j]))
            if diff < 1e-5:
                row.append((i, j))
        return row
    
    def get_non_tree_edges(self):
        self.non_tree_edges = []
        n = self.n
        tol = 1e-5
        
        # 获取树边集合（使用整数标识）
        tree_edges = self.T_edges
        
        for i in range(n):
            for j in range(n):
                # 统一使用整数标识
                is_tree_edge = (i, j) in tree_edges
                is_tight = abs(self.C[i, j] - (self.uu[i] + self.vv[j])) < tol
                
                if not is_tree_edge or not is_tight:
                    self.non_tree_edges.append((i, j))
        
        #print("here is non tree edges")
        
        # Visualize all non-basic edges
        #self.visualize_non_tree_edges()
        #return self.non_tree_edges

    def process_cluster_wrapper(self, args):
        return process_cluster(*args)
    #def process_cluster_wrapper(cluster):
        #return process_cluster(cluster, self.T_edges, self.n)
        
    def cluster_edges(self, edges):

        """使用DBSCAN进行高效聚类"""
        coords = np.array([(i, j) for (i, j) in edges])
        clustering = DBSCAN(eps=0.7*self.n, min_samples=1).fit(coords)
        
        edge_clusters = []
        for label in set(clustering.labels_):
            cluster = [edges[i] for i in np.where(clustering.labels_ == label)[0]]
            edge_clusters.append(cluster)
        
        return edge_clusters

    # 优化：基于几何位置剪枝
    def is_spatial_close(self, edge1, edge2, threshold):
        """检查两条边是否空间接近"""
        #threshold = 0.3 * self.n
        #threshold = self.spatial_threshold
        i1, j1 = edge1
        i2, j2 = edge2
        return abs(i1-i2) + abs(j1-j2) < threshold

    def find_valid_structures(self):
        """使用joblib并行处理"""
        from joblib import Parallel, delayed
        
        clusters = self.cluster_edges(self.non_tree_edges)
        self.valid_cycles = []
        
        # 并行处理
        with Parallel(n_jobs=4, prefer="threads") as parallel:
            results = parallel(
                delayed(process_cluster)(cluster, self.bipartite_edges, self.n)
                for cluster in clusters
            )
            for res in results:
                self.valid_cycles.extend(res)
        
        self.logger.info(f"找到有效环: {len(self.valid_cycles)}个")                     
        
    def compute_new_edges(self, cycle_edges):
        """
        计算通过环边替换后的新边集
        :param cycle_edges: 有效环的边集合
        :return: 新边集 (AP解与环边的对称差)
        """
        new_edges = set(self.ap_edges)  # 复制AP解边集
        for edge in cycle_edges:
            if edge in new_edges:
                new_edges.remove(edge)  # 移除已存在的边
            else:
                new_edges.add(edge)     # 添加新边
        return new_edges
    
    def generate_candidates(self):
        # 使用实时计算代替预存储
        best_cost = float('inf')
        best_tour = None
        
        # AP解
        if self.is_hamiltonian(self.ap_edges):
            cost = self.ap_cost
            if cost < best_cost:
                best_cost = cost
                best_tour = self.ap_edges
        
        # 有效环解
        for cycle_edges in self.valid_cycles:
            new_edges = self.compute_new_edges(cycle_edges)
            if self.is_hamiltonian(new_edges):
                cost = sum(self.C[i][j] for (i, j) in new_edges)
                if cost < best_cost:
                    best_cost = cost
                    best_tour = new_edges
        
        return best_tour, best_cost
    
    
    def find_optimal_tour(self):
        """找到最优ATSP解"""
        optimal_cost = float('inf')
        optimal_tour = None
        print("optimal tour YYC\n")
        print(optimal_tour)
        
        for edges, cost in self.candidates.items():
            if self.is_hamiltonian(edges) and cost < optimal_cost:
                optimal_cost = cost
                optimal_tour = edges
        
        return optimal_tour, optimal_cost
    
    def is_hamiltonian(self, edges):
        """高效检查Hamilton回路"""
        if len(edges) != self.n:  # 边数必须等于节点数
            return False
        
        graph = {}
        for i, j in edges:
            if i in graph:
                return False  # 出度>1
            graph[i] = j
        
        visited = set()
        node = 0
        for _ in range(self.n):
            visited.add(node)
            node = graph.get(node)
            if node is None:
                return False
        
        return len(visited) == self.n
    
    def visualize_matching(self, edges, title, subtitle=""):
        """可视化匹配（有向图版本）"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))  # 添加所有节点
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:  # 避免自环
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', style='dashed', alpha=0.3, arrowstyle='-|>', arrowsize=10)
    
        # 添加匹配边（红色实线）
        for i, j in edges:
            G.add_edge(i, j, color='red', width=2.0, arrowstyle='-|>', arrowsize=15)
    
        # 使用环形布局
        pos = nx.circular_layout(G)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(
            G, pos, edgelist=all_edges,
            edge_color='lightgray', width=0.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3
        )
    
        # 绘制匹配边（红色实线）
        match_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'red']
        nx.draw_networkx_edges(
            G, pos, edgelist=match_edges,
            edge_color='red', width=2.0, 
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 添加标题
        plt.title(f"{title}\n{subtitle}", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_spanning_tree(self, T):
        """可视化生成树（有向图版本）"""
        fig = plt.figure(figsize=(6, 5))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', style='dashed', alpha=0.3)
    
        # 添加生成树中的边（蓝色实线）
        tree_edges = set()
        for u, v in T.edges():
            # 转换二分图边为有向图边
            if u < self.n and v >= self.n:  # A->B边
                i, j = u, v - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
            elif v < self.n and u >= self.n:  # B->A边（转换为反向）
                i, j = v, u - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
    
        # 添加AP解中的边（红色实线）
        ap_edges = set()
        for i, j in self.ap_edges:
            ap_edges.add((i, j))
            if (i, j) not in tree_edges:  # 如果不在树中，添加为红色
                G.add_edge(i, j, color='red', width=2.0)
    
        # 使用弹簧布局
        pos = nx.spring_layout(G, k=0.5, iterations=50)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(G, pos, edgelist=all_edges, edge_color='lightgray', width=0.5, style='dashed',
        arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3)
    
        # 绘制生成树边（蓝色实线）
        tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'blue']
        nx.draw_networkx_edges(
            G, pos, edgelist=tree_edges,
            edge_color='blue', width=1.5, 
            arrows=True, arrowstyle='-|>', arrowsize=12)
    
        # 绘制AP解边（红色实线）
        ap_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'red']
        nx.draw_networkx_edges(
            G, pos, edgelist=ap_edges,
            edge_color='red', width=2.0, 
            arrows=True, arrowstyle='-|>', arrowsize=15)
    
        # 添加标题和图例
        plt.title(f"Spaning tree T* (contains AP solutions)", fontsize=14)
    
        # 创建图例
        import matplotlib.patches as mpatches
        gray_patch = mpatches.Patch(color='lightgray', label='possible edges')
        blue_patch = mpatches.Patch(color='blue', label='tree edges')
        red_patch = mpatches.Patch(color='red', label='AP solutions edges')
        plt.legend(handles=[gray_patch, blue_patch, red_patch], loc='best')
    
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_non_tree_edges(self):
        """可视化非树边（有向图版本）"""
        fig = plt.figure(figsize=(6, 5))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', 
                          style='dashed', alpha=0.3)
    
        # 添加树边（蓝色实线）
        tree_edges = set()
        for u, v in self.T_star.edges():
            # 转换二分图边为有向图边
            if u < self.n and v >= self.n:  # A->B边
                i, j = u, v - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
            elif v < self.n and u >= self.n:  # B->A边（转换为反向）
                i, j = v, u - self.n
                tree_edges.add((i, j))
                G.add_edge(i, j, color='blue', width=1.5)
    
        # 添加非树边（绿色虚线）
        for (i, j) in self.non_tree_edges:
            # 确保边不存在于树边中
            if (i, j) not in tree_edges:
                G.add_edge(i, j, color='green', width=2.0, style='dashed')
        
        # 使用弹簧布局
        pos = nx.spring_layout(G, k=0.5, iterations=50)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(
            G, pos, edgelist=all_edges,
            edge_color='lightgray', width=0.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3
        )
    
        # 绘制树边（蓝色实线）
        tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'blue']
        nx.draw_networkx_edges(
            G, pos, edgelist=tree_edges,
            edge_color='blue', width=1.5, 
            arrows=True, arrowstyle='-|>', arrowsize=12
        )
    
        # 绘制非树边（绿色虚线）
        non_tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'green']
        nx.draw_networkx_edges(
            G, pos, edgelist=non_tree_edges,
            edge_color='green', width=2.0, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 添加标题和图例
        plt.title("Spaning tree and non-tree edges", fontsize=14)
    
        # 创建图例
        import matplotlib.patches as mpatches
        gray_patch = mpatches.Patch(color='lightgray', label='possible edges')
        blue_patch = mpatches.Patch(color='blue', label='tree edges')
        green_patch = mpatches.Patch(color='green', label='non-tree edges')
        plt.legend(handles=[gray_patch, blue_patch, green_patch], loc='best')
    
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_cycle(self, edges, non_tree_set):
        """可视化有效环（有向图版本）"""
        fig = plt.figure(figsize=(2, 1))
        self.figures.append(fig)
    
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
    
        # 添加所有可能的边（灰色虚线）
        for i in range(self.n):
            for j in range(self.n):
                if i != j:
                    G.add_edge(i, j, weight=self.C[i][j], color='lightgray', 
                              style='dashed', alpha=0.3)
    
        # 添加环中的边
        for i, j in edges:
            if (i, j) in non_tree_set:
                # 非树边（绿色虚线）
                G.add_edge(i, j, color='green', width=2.5, style='dashed')
            else:
                # 树边（蓝色实线）
                G.add_edge(i, j, color='blue', width=2.0)
    
        # 使用环形布局（适合显示环）
        pos = nx.circular_layout(G)
    
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=500, node_color='skyblue')
    
        # 绘制标签
        labels = {i: f"{i}" for i in range(self.n)}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')
    
        # 绘制所有可能的边（灰色虚线）
        all_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'lightgray']
        nx.draw_networkx_edges(
            G, pos, edgelist=all_edges,
            edge_color='lightgray', width=0.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=10, alpha=0.3
        )
    
        # 绘制树边（蓝色实线）
        tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'blue']
        nx.draw_networkx_edges(
            G, pos, edgelist=tree_edges,
            edge_color='blue', width=2.0, 
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 绘制非树边（绿色虚线）
        non_tree_edges = [e for e in G.edges() if G[e[0]][e[1]].get('color') == 'green']
        nx.draw_networkx_edges(
            G, pos, edgelist=non_tree_edges,
            edge_color='green', width=2.5, style='dashed',
            arrows=True, arrowstyle='-|>', arrowsize=15
        )
    
        # 添加标题和图例
        non_tree_count = len([e for e in edges if e in self.non_tree_edges])
        plt.title(f"Valid cycles (non-tree edges: {non_tree_count})", fontsize=14)
    
        # 创建图例
        import matplotlib.patches as mpatches
        gray_patch = mpatches.Patch(color='lightgray', label='possbile edges')
        blue_patch = mpatches.Patch(color='blue', label='tree edges')
        green_patch = mpatches.Patch(color='green', label='non-tree edges')
        plt.legend(handles=[gray_patch, blue_patch, green_patch], loc='best')
    
        # 高亮显示环路径
        path_edges = []
        for i in range(len(edges)):
            path_edges.append(edges[i])
            if i < len(edges) - 1:
                # 添加中间节点
                pass
    
        # 添加环路径的动画效果（可选）
        for i, (u, v) in enumerate(path_edges):
            nx.draw_networkx_edges(
                G, pos, edgelist=[(u, v)],
                edge_color='red', width=3.0, 
                arrows=True, arrowstyle='-|>', arrowsize=20,
                alpha=0.7
            )
    
        plt.axis('off')
        plt.tight_layout()
    
    def visualize_tour(self, tour_edges, cost):
        """可视化ATSP环"""
        fig = plt.figure(figsize=(5, 4))
        self.figures.append(fig)
        
        # 创建有向图
        G = nx.DiGraph()
        G.add_nodes_from(range(self.n))
        
        # 添加边
        for i, j in tour_edges:
            G.add_edge(i, j, weight=self.C[i][j])
        
        # 创建环形布局
        pos = nx.circular_layout(G)
        
        # 绘制节点
        nx.draw_networkx_nodes(G, pos, node_size=700, node_color='lightgreen')
        nx.draw_networkx_labels(G, pos, font_size=12, font_weight='bold')
        
        # 绘制边
        edge_labels = {(i, j): f"{self.C[i][j]:.2f}" for i, j in G.edges()}
        nx.draw_networkx_edges(G, pos, edge_color='blue', width=2.0, 
                              arrows=True, arrowsize=20, 
                              connectionstyle='arc3,rad=0.1')
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10)
        
        # 添加标题
        plt.title(f"ATSP optimal solutions\ntotal costs: {cost:.4f}", fontsize=14)
        plt.axis('off')
        plt.tight_layout()
    
    def show_visualizations(self):
        """显示所有可视化结果"""
        print(f"显示 {len(self.figures)} 个可视化图表...")
        plt.show()

    def edges_to_tour(self):
        """
        将AP解边集转换为ATSP环路序列
        :param edges: 边集合，如 {(0,1), (1,2), (2,0)}
        :return: 环路序列列表，如 [0,1,2,0]
        """
        if not self.ap_edges:
            raise ValueError("边集不能为空")
    
        # 构建邻接字典：{起点: 终点}
        adj_map = {}
        for u, v in self.ap_edges:
            if u in adj_map:
                #raise ValueError(f"顶点 {u} 存在多条出边，非合法环路")
                return 0
            adj_map[u] = v
    
        # 验证顶点一致性
        all_vertices = set(adj_map.keys()) | set(adj_map.values())
        if len(all_vertices) != len(adj_map):
            #raise ValueError("存在孤立顶点或度数异常")
            return 0

        # 从任意起点开始追踪路径
        start = next(iter(adj_map.keys()))
        tour = [start]
        current = start
        visited = set([start])
    
        # 沿邻接映射追踪路径
        while True:
            next_node = adj_map.get(current)
            if next_node is None:
                #raise ValueError(f"顶点 {current} 无出边，路径中断")
                return 0
            if next_node in visited and next_node != start:
                #raise ValueError(f"检测到子环：顶点 {next_node} 被重复访问")
                return 0
        
            # 回到起点时结束
            if next_node == start:
                tour.append(start)
                break
            
            tour.append(next_node)
            visited.add(next_node)
            current = next_node

        # 验证路径覆盖所有顶点
        if len(visited) != len(all_vertices):
            missing = all_vertices - set(tour)
            #raise ValueError(f"未覆盖所有顶点，缺失顶点: {missing}")
            return 0

        print("hello tour")
        print(tour)
        self.logger.info("hello tour...")
        self.logger.info(tour)
        return 1
        
    
    def solve(self):
        """执行完整算法"""
        print(f"solve {self.n} cities ATSP problem (k_max = {self.k_max})")
        
        # 步骤1: 求解AP
        self.solve_ap()
        #print(f"AP optimal cost: {self.ap_cost:.4f}")
        
        cycle = self.edges_to_tour()
        if cycle == 1:
            results_true.append(1)
            return None, float('inf')
        if cycle == 0 and self.C.shape[0] > 151:
            results_true.append(0)
            percentage.append(0)
            return None, float('inf')
            
        if cycle == 0 and self.C.shape[0] <= 151:
            # step 2.1 get dual variables
            
            # 步骤2: 构建生成树T*
            self.build_spanning_tree()
            print(f"spaning tree T* construction completed ({len(self.T_edges)} edges)")
        
            # 步骤3: 识别非树边
            self.get_non_tree_edges()
            print(f"non-basic edges number: {len(self.non_tree_edges)}")
        
            # 步骤4: 枚举有效环
            self.find_valid_structures()
        
            # 步骤5: 生成候选解
            self.candidates = {}  # 重置候选解字典
            edges, cost = self.generate_candidates()
            if edges is not None:
                key = tuple(sorted(edges))
                self.candidates[key] = cost
            print(f"生成候选解数量: {len(self.candidates)}")    

            # 步骤6: 找到最优ATSP解
            tour_edges, cost = self.find_optimal_tour()
        
            if tour_edges:
                print(f"🎯 found ATSP optimal solution: cost = {cost:.4f}")
                self.visualize_tour(tour_edges, cost)
                results_true.append(1)
                return tour_edges, cost
            else:
                print("未找到Hamilton环解")
                results_true.append(0)
                percentage.append(0)
                return None, float('inf')
    
def process_cluster(cluster, bipartite_edges, n):
    """独立函数处理单个簇"""
    valid_cycles_local = []
    for k in range(1, min(2, len(cluster))):
        for non_tree_set in combinations(cluster, k):
            subG = nx.Graph()
            subG.add_edges_from(bipartite_edges)  # 从边列表重建图
            
            # 添加非树边
            for (i, j) in non_tree_set:  # 仅添加当前组合中的边
                subG.add_edge(i, j + n)
            
            try:
                for cycle in nx.simple_cycles(subG.to_directed()):
                    if len(cycle) % 2 == 0:  # 交替环必须偶数长度
                        edges = []
                        valid = True
                        
                        # 将节点序列转换为边序列
                        for idx in range(len(cycle)):
                            u = cycle[idx]
                            v = cycle[(idx + 1) % len(cycle)]
                            
                            if u < n and v >= n:  # A->B
                                edge = (u, v - n)
                            elif u >= n and v < n:  # B->A
                                edge = (v, u - n)
                            else:
                                valid = False
                                break
                            
                            edges.append(edge)
                        
                        if valid:
                            valid_cycles_local.append(edges)
            except nx.NetworkXNoCycle:
                pass
    return valid_cycles_local


def generate_random_instance(n, dist_type='exp'):
    """Generate random ATSP inistances"""
    if dist_type == 'exp':
        # exponential dist
        C = np.random.exponential(scale=1.0, size=(n, n))
    elif dist_type == 'uniform':
        # Uniform dist
        C = np.random.uniform(0, 1, size=(n, n))
    else:
        # random dist P(C≤x)=x+O(x²)
        C = np.zeros((n, n))
        for i in range(n):
            for j in range(n):
                u = np.random.uniform(0, 1)
                C[i][j] = u - 0.5 * u**2
    
    # execpt for the diagonal
    np.fill_diagonal(C, np.random.uniform(0, 1, n))
    return C


def verify_solution(C, tour):
    """Verify the correctness of the solution"""
    cost = 0
    visited = set()
    current = 0
    
    for _ in range(len(C)):
        next_node = tour[current]
        cost += C[current][next_node]
        visited.add(next_node)
        current = next_node
    
    return len(visited) == len(C), cost





# ===================== Main program =====================
def main(n):
    np.random.seed(42)
    
    #print("="*50)
    print("Bell & Frieze (2024) ATSP algorithm implementation")
    #print("="*50)
    
    # Generate a random instance (n=5)
    n = n
    print(f"\nGenrate a {n} cities random ATSP instance...")
    C = generate_random_instance(n, dist_type='uniform')
    for i in range(n):
        C[i][i] = 10
    #print("Cost matrix:")
    #print(np.round(C, 3))
    #C2 = C.tolist()
    #print(C2)
    
    # Generate a solver
    solver = BellFriezeATSP(C)
    
    # solve ATSP
    print("\nBegin to solve...")
    start = time.perf_counter()
    
    tour_edges, cost = solver.solve()

    end = time.perf_counter()
    timecount.append(end - start)
    del start, end
    
    if tour_edges:
        # transform the list of edges into a tour
        graph = {i: j for (i, j) in tour_edges}
        tour_path = [0]
        current = 0
        for _ in range(n-1):
            current = graph[current]
            tour_path.append(current)
        
        # Verify and print the tour
        valid, verified_cost = verify_solution(C, graph)
        print(f"Otpimal tour: {tour_path}")
        print(f"Verification: {'Sucess' if valid else 'failed'}, cost={verified_cost:.4f}")
        actual_cost = solve_tsp_from_matrix(C, show_plot=False)
        percentage.append(verified_cost/actual_cost)
        del actual_cost, verified_cost
        
        # Show all of the visualization
        #solver.show_visualizations()
```
