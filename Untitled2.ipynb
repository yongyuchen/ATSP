```python
import numpy as np
from scipy.optimize import linear_sum_assignment
from scipy.sparse.csgraph import connected_components
from scipy.sparse import csr_matrix
from pyscipopt import Model  # 添加SCIP求解器

def build_T_star(C):
    """
    构建生成树 T* 基于AP最优解
    参数:
        C: n x n 成本矩阵
    返回:
        T_star: 生成树边集（含匹配边和树边）
        M_star: 最优匹配
        u, v: 对偶变量
    """
    n = C.shape[0]
    tol = 1e-5  # 浮点精度容忍度

    # Step 1: 求解指派问题获取最优匹配
    row_ind, col_ind = linear_sum_assignment(C)
    M_star = list(zip(row_ind, col_ind))
    
    # Step 2: 使用SCIP求解LP获取精确对偶变量
    u, v = get_dual_variables_with_scip(C)
    
    # Step 3: 识别所有紧边（满足 C[i,j] ≈ u_i + v_j）
    tight_edges = []
    for i in range(n):
        for j in range(n):
            # 考虑浮点精度问题
            diff = abs(C[i, j] - (u[i] + v[j]))
            if diff < tol:
                tight_edges.append((i, j))
    
    # Step 4: 构建连通分量（初始为n个独立分量）
    parent = list(range(2 * n))  # 顶点索引: A:0~n-1, B:n~2n-1
    
    # 合并匹配边（每个匹配边连接A_i和B_j）
    for i, j in M_star:
        union(parent, i, j + n)
    
    # Step 5: 添加n-1条非匹配紧边连接分量
    T_edges = set(M_star)  # 初始化T*包含所有匹配边
    for (i, j) in tight_edges:
        if (i, j) in M_star:  # 跳过匹配边
            continue
            
        a_node = i
        b_node = j + n
        if find(parent, a_node) != find(parent, b_node):
            union(parent, a_node, b_node)
            T_edges.add((i, j))
            if len(T_edges) == 2 * n - 1:  # 已达2n-1条边
                break
    
    return list(T_edges), M_star, u, v

def get_dual_variables_with_scip(C):
    """
    使用SCIP求解指派问题的LP并获取对偶变量
    """
    n = C.shape[0]
    model = Model("AssignmentProblem")
    
    # 创建变量
    x = {}
    for i in range(n):
        for j in range(n):
            x[i, j] = model.addVar(vtype="C", lb=0, ub=1, name=f"x_{i}_{j}")
    
    # 设置目标函数：最小化总成本
    model.setObjective(sum(C[i, j] * x[i, j] for i in range(n) for j in range(n)), "minimize")
    
    # 添加行约束：每行和为1
    row_constraints = []
    for i in range(n):
        cons = model.addCons(sum(x[i, j] for j in range(n)) == 1, name=f"row_{i}")
        row_constraints.append(cons)
    
    # 添加列约束：每列和为1
    col_constraints = []
    for j in range(n):
        cons = model.addCons(sum(x[i, j] for i in range(n)) == 1, name=f"col_{j}")
        col_constraints.append(cons)
    
    # 求解模型
    model.optimize()
    
    # 检查求解状态
    if model.getStatus() != "optimal":
        raise RuntimeError("SCIP failed to find optimal solution")
    
    # 提取对偶变量
    u = np.zeros(n)  # 行约束的对偶变量
    v = np.zeros(n)  # 列约束的对偶变量
    
    for i in range(n):
        u[i] = model.getDualsolLinear(row_constraints[i])
    
    for j in range(n):
        v[j] = model.getDualsolLinear(col_constraints[j])
    
    return u, v

# --- 并查集工具函数 ---
def find(parent, x):
    if parent[x] != x:
        parent[x] = find(parent, parent[x])
    return parent[x]

def union(parent, x, y):
    root_x = find(parent, x)
    root_y = find(parent, y)
    if root_x != root_y:
        parent[root_y] = root_x

# 示例使用
if __name__ == "__main__":
    # 示例成本矩阵（n=3）
    C = np.array([
        [0.1, 0.6, 0.3],
        [0.2, 0.8, 0.4],
        [0.5, 0.9, 0.7]
    ])
    
    T_star, M_star, u, v = build_T_star(C)
    
    print("最优匹配 M*:", M_star)
    print("对偶变量 u:", u)
    print("对偶变量 v:", v)
    print("生成树 T* 边集（格式: (A_i, B_j):")
    for edge in T_star:
        print(f"({edge[0]}, {edge[1]})")
```

```python
import numpy as np
from pyscipopt import Model
import networkx as nx
import time
import itertools
from collections import defaultdict

class BellFriezeATSP:
    def __init__(self, cost_matrix):
        self.C = cost_matrix
        self.n = cost_matrix.shape[0]
        self.ap_edges = []  # 存储最优匹配边
        self.u = None  # 对偶变量u
        self.v = None  # 对偶变量v
        self.T_star = None  # 生成树
        self.non_tree_edges = []  # 非树边集合
        
    def solve(self):
        """主求解函数"""
        start_time = time.time()
        
        # 步骤1: 求解指派问题并获取对偶变量
        self.solve_assignment_problem()
        print("Assignment problem solved")
        
        # 步骤2: 构建生成树T*
        self.build_spanning_tree()
        print(f"Spanning tree T* constructed with {len(self.T_star.edges())} edges")
        
        # 步骤3: 识别非树边
        self.identify_non_tree_edges()
        print(f"Identified {len(self.non_tree_edges)} non-tree edges")
        
        # 步骤4: 寻找最优环游
        tour_edges, cost = self.find_optimal_tour()
        
        print(f"Total time: {time.time() - start_time:.4f} seconds")
        return tour_edges, cost
    
    def solve_assignment_problem(self):
        """使用SCIP求解指派问题并获取对偶变量"""
        n = self.n
        model = Model("AssignmentProblem")
        
        # 创建变量
        x = {}
        for i in range(n):
            for j in range(n):
                x[i, j] = model.addVar(vtype="C", lb=0, ub=1, name=f"x_{i}_{j}")
        
        # 设置目标函数：最小化总成本
        model.setObjective(sum(self.C[i, j] * x[i, j] for i in range(n) for j in range(n)), "minimize")
        
        # 添加行约束：每行和为1
        row_constraints = []
        for i in range(n):
            cons = model.addCons(sum(x[i, j] for j in range(n)) == 1, name=f"row_{i}")
            row_constraints.append(cons)
        
        # 添加列约束：每列和为1
        col_constraints = []
        for j in range(n):
            cons = model.addCons(sum(x[i, j] for i in range(n)) == 1, name=f"col_{j}")
            col_constraints.append(cons)
        
        # 求解模型
        model.optimize()
        
        # 检查求解状态
        if model.getStatus() != "optimal":
            raise RuntimeError("SCIP failed to find optimal solution")
        
        # 保存最优匹配边
        self.ap_edges = []
        for i in range(n):
            for j in range(n):
                if model.getVal(x[i, j]) > 0.5:  # 大于0.5视为选中
                    self.ap_edges.append((i, j))
        
        # 提取对偶变量
        self.u = np.zeros(n)
        self.v = np.zeros(n)
        
        for i in range(n):
            self.u[i] = model.getDualsolLinear(row_constraints[i])
        
        for j in range(n):
            self.v[j] = model.getDualsolLinear(col_constraints[j])
    
    def build_spanning_tree(self):
        """构建生成树T*"""
        n = self.n
        tol = 1e-5
        
        # 步骤1: 识别所有紧边（满足 C[i,j] ≈ u_i + v_j）
        tight_edges = []
        for i in range(n):
            for j in range(n):
                diff = abs(self.C[i, j] - (self.u[i] + self.v[j]))
                if diff < tol:
                    tight_edges.append((i, j))
        
        # 步骤2: 初始化并查集
        parent = list(range(2 * n))  # 顶点索引: A:0~n-1, B:n~2n-1
        
        # 步骤3: 合并匹配边
        for i, j in self.ap_edges:
            self.union(parent, i, j + n)
        
        # 步骤4: 添加非匹配紧边连接分量
        T_edges = set(self.ap_edges)
        for (i, j) in tight_edges:
            if (i, j) in self.ap_edges:  # 跳过匹配边
                continue
                
            a_node = i
            b_node = j + n
            if self.find(parent, a_node) != self.find(parent, b_node):
                self.union(parent, a_node, b_node)
                T_edges.add((i, j))
                if len(T_edges) == 2 * n - 1:  # 已达2n-1条边
                    break
        
        # 构建生成树图
        self.T_star = nx.Graph()
        for (i, j) in T_edges:
            self.T_star.add_edge(f"A_{i}", f"B_{j}", weight=self.C[i, j])
        print("T_starskfjasjflajfaj")
        A = nx.to_numpy_array(self.T_star)
        print(A)
    
    def identify_non_tree_edges(self):
        """识别所有非树边"""
        self.non_tree_edges = []
        n = self.n
        tol = 1e-5
        
        # 获取树边集合
        tree_edges = set(self.T_star.edges())
        
        for i in range(n):
            for j in range(n):
                # 检查是否为树边
                is_tree_edge = (f"A_{i}", f"B_{j}") in tree_edges or (f"B_{j}", f"A_{i}") in tree_edges
                
                # 检查是否满足紧边条件
                is_tight = abs(self.C[i, j] - (self.u[i] + self.v[j])) < tol
                
                # 非树边或非紧边
                if not is_tree_edge or not is_tight:
                    self.non_tree_edges.append((i, j))
    
    def find_optimal_tour(self):
        """寻找最优环游（简化版）"""
        # 在实际实现中，这里应包含论文中的完整搜索算法
        # 此处简化：直接使用最近邻算法作为示例
        return self.nearest_neighbor()
    
    def nearest_neighbor(self):
        """最近邻启发式算法作为示例"""
        n = self.n # size of the nodes
        unvisited = set(range(n)) # all nodes are unvisited
        current = 0
        tour = [current] # set 0 as the first node of the tour
        total_cost = 0.0 # set cost to be 0
        
        while unvisited: # remove current node from the unvisited
            unvisited.remove(current)
            if not unvisited:
                break
                
            # 寻找最近邻居
            nearest = None
            min_cost = float('inf')
            for neighbor in unvisited: # select the minimal cost neighbor
                cost = self.C[current, neighbor]
                if cost < min_cost:
                    min_cost = cost
                    nearest = neighbor
            
            total_cost += min_cost
            current = nearest
            tour.append(current)
        
        # 返回起点形成环
        total_cost += self.C[tour[-1], tour[0]]
        tour_edges = [(tour[i], tour[i+1]) for i in range(len(tour)-1)]
        tour_edges.append((tour[-1], tour[0]))
        
        return tour_edges, total_cost
    
    # 并查集工具函数
    @staticmethod
    def find(parent, x):
        """查找根节点（带路径压缩）"""
        if parent[x] != x:
            parent[x] = BellFriezeATSP.find(parent, parent[x])
        return parent[x]
    
    @staticmethod
    def union(parent, x, y):
        """合并两个集合"""
        root_x = BellFriezeATSP.find(parent, x)
        root_y = BellFriezeATSP.find(parent, y)
        if root_x != root_y:
            parent[root_y] = root_x


# 测试函数
def test_atsp(n=5):
    # 生成随机成本矩阵
    np.random.seed(42)
    C = np.random.rand(n, n) * 10
    np.fill_diagonal(C, 100)  # 对角线设为0
    
    print("Cost matrix:")
    print(C)
    
    # 创建求解器实例
    solver = BellFriezeATSP(C)
    
    # 求解问题
    tour_edges, cost = solver.solve()
    
    print("\nOptimal tour edges:", tour_edges)
    print("Tour cost:", cost)
    
    # 验证环游
    if tour_edges:
        nodes = set()
        for edge in tour_edges:
            nodes.add(edge[0])
            nodes.add(edge[1])
        
        if len(nodes) == n and len(tour_edges) == n:
            print("Valid tour: visits all nodes")
        else:
            print("Invalid tour")

if __name__ == "__main__":
    test_atsp(5)
```
